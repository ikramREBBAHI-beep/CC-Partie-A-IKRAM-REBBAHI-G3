## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression linéaire** (*Linear Regression*, moindres carrés ordinaires) — famille : **régression supervisée**. Objectif : prédire une cible numérique continue à partir de variables explicatives.
- **Intuition et étapes** : on suppose que la cible est une combinaison pondérée des variables. Entrée : matrice $X$ ($n\times p$) et cible $y$ → apprentissage : on cherche les coefficients qui minimisent la somme des carrés des erreurs → sortie : une prédiction $\hat y$ pour toute nouvelle observation.
- **Modèle et critère** : $\hat y_i=\beta_0+\sum_{j=1}^{p}\beta_j x_{ij}$ ; $J(\beta)=\frac1{2n}\sum_{i=1}^n (y_i-\hat y_i)^2$. Solution fermée : $\hat\beta=(X^\top X)^{-1}X^\top y$.
- **Symboles** : $n$ nombre d'observations, $p$ nombre de variables, $x_{ij}$ valeur de la variable $j$ pour l'observation $i$, $y_i$ cible observée, $\beta_0$ constante, $\beta_j$ coefficient.
- **Paramètres appris** : $\beta_0,\dots,\beta_p$. **Hyperparamètre** : pratiquement aucun (seulement `fit_intercept`).
- **Avantages** : très rapide, coefficients directement interprétables (effet *toutes choses égales par ailleurs*).
- **Limites** : sensible aux valeurs extrêmes et à la colinéarité ; ne capte pas les relations non linéaires sans transformation des variables.

### Schéma explicatif

**Comment lire le schéma :** les points gris sont les observations, la droite rouge est la prédiction du modèle et chaque segment vertical est un **résidu** $y_i-\hat y_i$. L'algorithme choisit la droite qui rend la somme des carrés de ces segments la plus petite possible.

```python
m = LinearRegression().fit(x1.reshape(-1,1), y1); pred = m.predict(x1.reshape(-1,1))
plt.figure(figsize=(6,3.6)); plt.scatter(x1,y1,s=14,c='gray',label='observations (synthétiques)')
plt.plot(xg,m.predict(xg.reshape(-1,1)),'r',lw=2,label='droite ajustée')
plt.vlines(x1,y1,pred,colors='tab:blue',lw=.8,label='résidus')
plt.title('Régression linéaire : minimiser la somme des résidus au carré'); plt.xlabel('x'); plt.ylabel('y'); plt.legend(fontsize=8); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (finance/gestion)** : prévoir le chiffre d'affaires d'un point de vente à partir du trafic, du prix moyen et du budget marketing.
- **Cas 2 (autre domaine)** : estimer le prix d'un logement à partir de sa surface et de son nombre de pièces.
- **Données et prérequis** : cible numérique, variables numériques (ou catégorielles encodées), relation approximativement linéaire, peu de colinéarité, résidus sans structure.
- **Peu approprié** : prévoir une demande avec saisonnalité marquée et effets de seuil (relation en escalier ou cyclique) — une droite ne peut pas représenter ces formes.
- **Métriques** : MAE, RMSE, R² sur un jeu de test ; analyse des résidus.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression Ridge** (*Ridge Regression*, régularisation L2) — famille : **régression supervisée**. Objectif : prédire une cible numérique en **stabilisant** les coefficients quand les variables sont corrélées ou nombreuses.
- **Intuition et étapes** : on ajoute à l'erreur quadratique une pénalité qui punit les gros coefficients. Entrée : $X$ standardisé et $y$ → apprentissage : minimisation de l'erreur + pénalité → sortie : prédiction plus stable que les moindres carrés.
- **Critère** : $J(\beta)=\frac1n\lVert y-X\beta\rVert_2^2+\lambda\sum_{j=1}^p\beta_j^2$ (dans scikit-learn, `alpha` joue le rôle de $\lambda$, avec une normalisation différente du terme d'erreur).
- **Symboles** : $X$ matrice $n\times p$, $y$ cible, $\beta$ vecteur des coefficients (constante non pénalisée), $\lambda\ge 0$ force de la pénalité, $\lVert\cdot\rVert_2$ norme euclidienne.
- **Paramètres appris** : $\beta$. **Hyperparamètre** : $\lambda$ (`alpha`), choisi sur validation.
- **Avantages** : stable face à la colinéarité ; solution unique, calcul rapide.
- **Limites** : ne met jamais exactement des coefficients à zéro (pas de sélection de variables) ; nécessite de standardiser les variables.

### Schéma explicatif

**Comment lire le schéma :** chaque courbe est un coefficient $\beta_j$ en fonction de $\alpha$ (axe horizontal logarithmique). Quand $\alpha$ augmente, tous les coefficients se rapprochent de 0 **sans l'atteindre** : c'est le « rétrécissement » de Ridge.

```python
Xs = StandardScaler().fit_transform(X_reg); alphas = np.logspace(-2,5,60)
coefs = np.array([Ridge(alpha=a).fit(Xs,y_reg).coef_ for a in alphas])
plt.figure(figsize=(6,3.6)); plt.semilogx(alphas,coefs)
plt.axhline(0,c='k',lw=.8); plt.title('Ridge : les coefficients rétrécissent quand α augmente')
plt.xlabel('α (échelle log)'); plt.ylabel('coefficient'); plt.legend([f'variable {j}' for j in range(coefs.shape[1])],fontsize=7); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (finance)** : prévoir le rendement d'un portefeuille avec de nombreux indicateurs macroéconomiques fortement corrélés (taux, inflation, indices).
- **Cas 2** : prédire une consommation d'énergie à partir de capteurs météo redondants (température, ressenti, point de rosée).
- **Données et prérequis** : variables numériques standardisées, éventuellement corrélées, $p$ proche de $n$ possible ; $\lambda$ choisi par validation croisée.
- **Peu approprié** : quand on veut **identifier un petit nombre de variables utiles** (Ridge les garde toutes) ou quand la relation est fortement non linéaire.
- **Métriques** : MAE, RMSE, R² ; stabilité des coefficients entre échantillons.

### Expérience Python fournie (code conservé)

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées** : mêmes données que le modèle 1, mais passées dans un `Pipeline` avec `StandardScaler` (le scaler est ajusté **uniquement** sur l'entraînement : pas de fuite). **Hyperparamètre** : `alpha=1.0`, fixé a priori et non optimisé.
- **Sortie** : MAE sur le test, à comparer à celle de la régression linéaire.
- **Conclusion possible** : avec 5 variables peu corrélées et 210 observations, il y a très peu de variance à réduire ; on s'attend donc à un résultat **proche** de la régression linéaire. Un écart minime n'est pas un échec de Ridge.
- **Non conclusif** : la valeur de `alpha` n'est pas validée ; pour conclure sur l'intérêt de Ridge il faudrait une recherche sur grille avec validation croisée (faite en partie B sur données réelles).

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression Lasso** (*Lasso*, régularisation L1) — famille : **régression supervisée**. Objectif : prédire une cible tout en **sélectionnant automatiquement** les variables utiles.
- **Intuition et étapes** : la pénalité porte sur la valeur absolue des coefficients ; elle a un « coin » en 0 qui pousse certains coefficients exactement à zéro. Entrée : $X$ standardisé, $y$ → minimisation (descente par coordonnées) → sortie : modèle parcimonieux.
- **Critère** : $J(\beta)=\frac1{2n}\lVert y-X\beta\rVert_2^2+\lambda\sum_{j=1}^{p}\lvert\beta_j\rvert$.
- **Symboles** : $n$ taille de l'échantillon, $X$ variables, $y$ cible, $\beta_j$ coefficient de la variable $j$, $\lambda$ intensité de la pénalité (`alpha`).
- **Paramètres appris** : $\beta$ (dont beaucoup peuvent valoir 0). **Hyperparamètres** : $\lambda$, `max_iter`.
- **Avantages** : sélection de variables intégrée ; modèle plus lisible.
- **Limites** : instable quand des variables sont très corrélées (il en retient une au hasard) ; biais sur les coefficients conservés.

### Schéma explicatif

**Comment lire le schéma :** en allant de gauche à droite, $\alpha$ augmente et les courbes **tombent à zéro l'une après l'autre**. Une variable dont la courbe est nulle est éliminée du modèle ; la dernière à disparaître est la plus influente.

```python
Xs = StandardScaler().fit_transform(X_reg); alphas = np.logspace(-2,2.3,60)
coefs = np.array([Lasso(alpha=a,max_iter=10000).fit(Xs,y_reg).coef_ for a in alphas])
plt.figure(figsize=(6,3.6)); plt.semilogx(alphas,coefs)
plt.axhline(0,c='k',lw=.8); plt.title('Lasso : des coefficients exactement nuls apparaissent')
plt.xlabel('α (échelle log)'); plt.ylabel('coefficient'); plt.legend([f'variable {j}' for j in range(coefs.shape[1])],fontsize=7); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (finance)** : sélectionner, parmi une centaine de ratios comptables, ceux qui expliquent le mieux la rentabilité d'une entreprise.
- **Cas 2** : repérer les quelques gènes ou capteurs qui pilotent une mesure quand $p\gg n$.
- **Données et prérequis** : variables standardisées, hypothèse de **parcimonie** (peu de variables réellement utiles), $\lambda$ choisi par validation croisée.
- **Peu approprié** : variables toutes informatives et très corrélées (groupes de prédicteurs quasi identiques) : la sélection devient arbitraire ; préférer Ridge ou Elastic Net.
- **Métriques** : R², MAE, RMSE, **nombre de coefficients non nuls**, stabilité de la sélection entre rééchantillonnages.

### Expérience Python fournie (code conservé)

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées** : mêmes données, pipeline `StandardScaler` + `Lasso`. **Hyperparamètres** : `alpha=0.1`, `max_iter=10000` (pour garantir la convergence).
- **Sortie** : le **R²** sur le test, c'est-à-dire la part de variance de la cible expliquée hors échantillon (1 = parfait, 0 = équivalent à prédire la moyenne).
- **Conclusion possible** : un R² élevé indique que le modèle capte l'essentiel du signal linéaire. Comme la sortie n'affiche que le R², **on ne voit pas** quelles variables ont été éliminées : il faudrait inspecter `modele[-1].coef_`.
- **Non conclusif** : `alpha=0.1` est faible, donc peu de sélection ; ici les 5 variables sont toutes informatives par construction, donc Lasso n'a pas d'avantage attendu sur la régression linéaire.

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression polynomiale** (*Polynomial Regression*) — famille : **régression supervisée**. Objectif : modéliser des relations **courbes** en restant dans un cadre linéaire.
- **Intuition et étapes** : on enrichit les variables avec leurs puissances et produits ($x,x^2,x^3,x_1x_2,\dots$), puis on applique une régression linéaire (ici pénalisée par Ridge) dans cet espace élargi.
- **Modèle** : $\hat y=\beta_0+\beta_1x+\beta_2x^2+\dots+\beta_dx^d$ — linéaire **dans les coefficients**, polynomial dans les variables.
- **Symboles** : $x$ variable d'entrée, $d$ degré maximal, $\beta_j$ coefficients appris, $\hat y$ valeur estimée.
- **Paramètres appris** : les $\beta_j$. **Hyperparamètres** : le degré $d$ (`degree`), la pénalité `alpha` du Ridge associé.
- **Avantages** : capte des courbures simples avec un modèle connu ; reste interprétable à faible degré.
- **Limites** : le nombre de termes explose avec $p$ et $d$ ; un degré trop élevé provoque du surapprentissage et des oscillations aux bords.

### Schéma explicatif

**Comment lire le schéma :** le degré 1 est trop rigide (sous-apprentissage), le degré 12 colle au bruit et oscille aux extrémités (surapprentissage) ; le degré intermédiaire suit la tendance sans suivre le bruit.

```python
from sklearn.preprocessing import PolynomialFeatures
courbes({f'degré {d}': make_pipeline(PolynomialFeatures(d,include_bias=False),StandardScaler(),LinearRegression()) for d in (1,3,12)},
        'Régression polynomiale : effet du degré')
```

### Contextes d'utilisation
- **Cas 1 (marketing/gestion)** : modéliser les ventes en fonction du budget publicitaire, avec rendements décroissants (courbe concave).
- **Cas 2** : relation température–consommation d'électricité (courbe en U : chauffage l'hiver, climatisation l'été).
- **Données et prérequis** : peu de variables, relation lisse, variables standardisées avant la création des puissances, degré choisi sur validation.
- **Peu approprié** : beaucoup de variables (explosion des termes) ou extrapolation hors du domaine d'apprentissage : les polynômes divergent rapidement.
- **Métriques** : MAE, RMSE, R² sur validation/test ; courbes d'apprentissage selon le degré.

### Expérience Python fournie (code conservé)

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),PolynomialFeatures(degree=2,include_bias=False),Ridge(alpha=10)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées** : pipeline `StandardScaler` → `PolynomialFeatures(degree=2)` → `Ridge(alpha=10)`. Avec 5 variables, le degré 2 crée 20 termes (5 carrés, 10 produits, 5 linéaires).
- **Hyperparamètres** : `degree=2`, `alpha=10` ; la pénalité limite le surapprentissage dû aux termes ajoutés.
- **Sortie** : MAE sur le test, à rapprocher de celle du modèle 1.
- **Conclusion possible** : la cible synthétique est **purement linéaire** ; les termes quadratiques n'apportent donc aucune information et on s'attend à une MAE égale ou légèrement supérieure à celle de la régression linéaire.
- **Non conclusif** : un seul degré testé ; la comparaison n'a de sens que si le degré est choisi sur un jeu de validation.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Arbre de régression** (*Regression Tree*, CART) — famille : **régression supervisée**. Objectif : prédire une valeur numérique par une suite de règles « si… alors… ».
- **Intuition et étapes** : l'algorithme teste des variables et des seuils, retient la coupure qui réduit le plus la dispersion de la cible, puis recommence dans chaque sous-groupe jusqu'à une profondeur maximale ; la prédiction d'une feuille est la **moyenne** de ses observations.
- **Critère de coupure** : choisir $(j,s)$ minimisant $\sum_{i\in G}(y_i-\bar y_G)^2+\sum_{i\in D}(y_i-\bar y_D)^2$ ; prédiction : $\hat y(x)=\frac1{|L(x)|}\sum_{i\in L(x)}y_i$.
- **Symboles** : $j$ variable testée, $s$ seuil, $G$/$D$ groupes gauche/droite, $L(x)$ feuille contenant $x$, $|L(x)|$ son effectif.
- **Paramètres appris** : variables et seuils de coupure, valeurs des feuilles. **Hyperparamètres** : `max_depth`, `min_samples_leaf`.
- **Avantages** : non linéaire, aucune standardisation nécessaire, règles lisibles.
- **Limites** : instable (un petit changement de données change l'arbre) ; ne sait pas extrapoler hors des valeurs vues.

### Schéma explicatif

**Comment lire le schéma :** la prédiction est une **fonction en escalier** : chaque palier est la moyenne d'une feuille. Un arbre peu profond (2) donne peu de marches ; un arbre profond (6) en donne beaucoup et commence à suivre le bruit.

```python
from sklearn.tree import DecisionTreeRegressor
courbes({'profondeur 2': DecisionTreeRegressor(max_depth=2), 'profondeur 6': DecisionTreeRegressor(max_depth=6)},
        'Arbre de régression : prédiction en escalier')
```

### Contextes d'utilisation
- **Cas 1 (logistique)** : estimer le coût de livraison selon la distance, le poids et le mode de transport, avec des effets de seuil (tranches tarifaires).
- **Cas 2** : prédire le prix d'un bien avec des règles de segmentation expliquables à un client.
- **Données et prérequis** : tableaux mixtes (numériques/catégorielles encodées), valeurs manquantes à traiter, profondeur contrôlée.
- **Peu approprié** : relation lisse et linéaire (l'escalier approxime mal une droite) ou prévision au-delà des valeurs observées.
- **Métriques** : MAE, RMSE, R² ; profondeur et nombre de feuilles pour juger la complexité.

### Expérience Python fournie (code conservé)

```python
from sklearn.tree import DecisionTreeRegressor
modele=DecisionTreeRegressor(max_depth=5,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées** : mêmes données ; `DecisionTreeRegressor` sans standardisation (inutile pour un arbre). **Hyperparamètres** : `max_depth=5`, `random_state=42` (le hasard sert à départager des coupures équivalentes).
- **Sortie** : MAE sur le test.
- **Conclusion possible** : comme la vraie relation est linéaire, un arbre de profondeur 5 l'approxime par paliers et on s'attend à une erreur **plus élevée** que celle des modèles linéaires. C'est une illustration du fait qu'un modèle non linéaire n'est pas toujours meilleur.
- **Non conclusif** : la profondeur n'est pas validée ; le résultat dépend du découpage (instabilité de l'arbre).

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Forêt aléatoire pour la régression** (*Random Forest Regressor*) — famille : **régression supervisée** (méthode d'ensemble par *bagging*). Objectif : réduire la variance d'un arbre en moyennant beaucoup d'arbres différents.
- **Intuition et étapes** : (1) tirer $B$ échantillons *bootstrap* des observations ; (2) entraîner un arbre sur chacun, en ne considérant à chaque coupure qu'un sous-ensemble aléatoire de variables ; (3) prédire la **moyenne** des prédictions des arbres.
- **Formule** : $\hat y(x)=\frac1B\sum_{b=1}^{B}T_b(x)$.
- **Symboles** : $B$ nombre d'arbres, $T_b(x)$ prédiction du $b$-ième arbre au point $x$, $\hat y$ prédiction finale.
- **Paramètres appris** : les $B$ arbres. **Hyperparamètres** : `n_estimators` ($B$), `max_depth`, `max_features`, `min_samples_leaf`.
- **Avantages** : robuste, performant sans gros réglage, gère les non-linéarités et interactions ; estimation d'erreur *out-of-bag*.
- **Limites** : modèle volumineux et peu interprétable ; extrapole mal hors du domaine observé.

### Schéma explicatif

**Comment lire le schéma :** les courbes fines sont des arbres individuels entraînés sur des échantillons bootstrap : ils fluctuent beaucoup. La courbe épaisse est leur **moyenne** : bien plus lisse et stable. C'est la réduction de variance du bagging.

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Gradient Boosting pour la régression** (*Gradient Boosting Regressor*, ici `HistGradientBoostingRegressor`) — famille : **régression supervisée** (ensemble séquentiel). Objectif : construire un modèle précis en corrigeant progressivement ses erreurs.
- **Intuition et étapes** : (1) initialiser avec une prédiction simple (la moyenne) ; (2) calculer les **résidus** (erreurs restantes) ; (3) entraîner un petit arbre pour les prédire ; (4) ajouter cet arbre au modèle, pondéré par un taux d'apprentissage ; répéter.
- **Formule** : $F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$, où $h_m$ approche l'opposé du gradient de la perte.
- **Symboles** : $F_m$ modèle à l'itération $m$, $h_m$ nouvel arbre (apprenant faible), $\eta$ taux d'apprentissage, $x$ observation.
- **Paramètres appris** : les arbres $h_m$. **Hyperparamètres** : `max_iter` (nombre d'arbres), `learning_rate` ($\eta$), `max_depth`.
- **Avantages** : très performant sur données tabulaires ; la version *Hist* est rapide (variables discrétisées en classes).
- **Limites** : réglage sensible, risque de surapprentissage si trop d'itérations ; moins interprétable.

### Schéma explicatif

**Comment lire le schéma :** de gauche à droite, le modèle passe de 1 à 5 puis à 50 arbres ajoutés. Avec peu d'arbres il reste grossier ; chaque arbre corrige une partie de l'erreur et la courbe se rapproche de la tendance.

- **Données et prérequis** : tableaux structurés de taille moyenne ou grande ; réglage par validation (nombre d'itérations, taux d'apprentissage).
- **Peu approprié** : très petits échantillons (surapprentissage), images/textes bruts, ou besoin d'explications simples.
- **Métriques** : MAE, RMSE, R² ; courbe d'erreur de validation selon le nombre d'itérations.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Régression logistique** (*Logistic Regression*) — famille : **classification supervisée**. Objectif : estimer la probabilité qu'une observation appartienne à la classe 1.
- **Intuition et étapes** : on calcule un score linéaire $z$, puis on le transforme en probabilité par la fonction sigmoïde ; on prédit la classe 1 si la probabilité dépasse un seuil (0,5 par défaut).
- **Modèle et critère** : $P(y=1\mid x)=\sigma(z)=\dfrac1{1+e^{-z}}$ avec $z=\beta_0+\beta^\top x$ ; perte (log-loss) $-\frac1n\sum_i\big[y_i\ln p_i+(1-y_i)\ln(1-p_i)\big]$.
- **Symboles** : $x$ variables, $\beta$ coefficients, $\sigma$ sigmoïde, $p_i$ probabilité prédite pour $i$, $y_i\in\{0,1\}$ classe observée.
- **Paramètres appris** : $\beta$. **Hyperparamètres** : $C$ (inverse de la régularisation), `penalty`, seuil de décision.
- **Avantages** : probabilités calibrées, coefficients interprétables (odds-ratios), rapide.
- **Limites** : frontière de décision linéaire ; sensible aux variables non standardisées et aux classes déséquilibrées (seuil à ajuster).

### Schéma explicatif

**Comment lire le schéma :** les points à 0 ou 1 sont les classes observées en fonction d'une variable ; la courbe en S est la probabilité estimée. La ligne pointillée horizontale est le seuil 0,5 : à droite de leur intersection, on prédit la classe 1.

```python
from sklearn.linear_model import LogisticRegression
r = np.random.default_rng(GRAINE); xa = r.normal(0,1.2,120); ya = (xa+r.normal(0,.8,120)>0).astype(int)
m = LogisticRegression().fit(xa.reshape(-1,1),ya); g = np.linspace(-4,4,200)
plt.figure(figsize=(6,3.4)); plt.scatter(xa,ya,c=ya,cmap='coolwarm',edgecolor='k',s=18,label='classes observées (synthétiques)')
plt.plot(g,m.predict_proba(g.reshape(-1,1))[:,1],'k',lw=2,label='P(y=1|x)'); plt.axhline(.5,ls='--',c='gray',label='seuil 0,5')
plt.title('Régression logistique : la sigmoïde transforme un score en probabilité'); plt.xlabel('x'); plt.ylabel('probabilité'); plt.legend(fontsize=8); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (finance)** : estimer la probabilité de défaut d'un client à partir de son historique de paiement (*credit scoring*).
- **Cas 2** : prédire si un client va résilier un abonnement.
- **Données et prérequis** : cible binaire, variables numériques standardisées ou encodées, taille suffisante pour la classe rare.
- **Peu approprié** : frontières très non linéaires sans variables d'interaction ; classes quasi parfaitement séparables (coefficients qui divergent sans régularisation).
- **Métriques** : précision, rappel, F1, ROC-AUC, PR-AUC si classe rare, calibration.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**K plus proches voisins** (*k-Nearest Neighbors*, KNN) — famille : **classification supervisée** (méthode à base d'instances, applicable aussi à la régression). Objectif : classer une observation selon ses voisins les plus proches.
- **Intuition et étapes** : « dis-moi qui sont tes voisins, je te dirai qui tu es ». Il n'y a pas d'apprentissage explicite : on mémorise les données, puis pour un nouveau point on calcule les distances, on prend les $k$ plus proches et on retient la classe majoritaire.
- **Formule** : $d(x,x')=\lVert x-x'\rVert_2$ ; $\hat y(x)=\text{classe majoritaire de }\{y_i : x_i\in\mathcal N_k(x)\}$.
- **Symboles** : $x$ point à classer, $x'$ point d'entraînement, $\mathcal N_k(x)$ ensemble des $k$ voisins les plus proches, $d$ distance euclidienne.
- **Paramètres appris** : aucun (les données elles-mêmes). **Hyperparamètres** : $k$, métrique de distance, pondération des voisins.
- **Avantages** : très simple, frontière flexible, aucune hypothèse de forme.
- **Limites** : lent en prédiction sur gros jeux ; très sensible à l'échelle des variables et à la dimension élevée (malédiction de la dimension).

### Schéma explicatif

**Comment lire le schéma :** à gauche, $k=1$ : la frontière épouse chaque point, y compris le bruit (variance forte). À droite, $k=25$ : la frontière est plus lisse (biais plus fort). $k$ règle le compromis biais-variance.

- **Données et prérequis** : variables numériques **standardisées**, effectif modéré, peu de dimensions, distance pertinente.
- **Peu approprié** : très grandes bases en temps réel ou centaines de variables (distances peu discriminantes, prédiction coûteuse).

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Naïve Bayes gaussien** (*Gaussian Naive Bayes*) — famille : **classification supervisée** (modèle probabiliste génératif). Objectif : classer en appliquant la règle de Bayes avec une hypothèse d'indépendance entre variables.
- **Intuition et étapes** : pour chaque classe, on estime la moyenne et la variance de chaque variable ; pour un nouveau point, on multiplie la probabilité a priori de la classe par la vraisemblance de chaque variable et on choisit la classe la plus probable.
- **Formule** : $P(c\mid x)\propto P(c)\prod_{j=1}^{p}\mathcal N(x_j;\mu_{cj},\sigma_{cj}^2)$.
- **Symboles** : $c$ classe, $P(c)$ proportion de la classe, $\mu_{cj},\sigma^2_{cj}$ moyenne et variance de la variable $j$ dans la classe $c$, $\mathcal N$ densité gaussienne.
- **Paramètres appris** : $P(c),\mu_{cj},\sigma_{cj}^2$. **Hyperparamètre** : `var_smoothing` (lissage de la variance).
- **Avantages** : entraînement quasi instantané, fonctionne avec peu de données.
- **Limites** : hypothèse d'indépendance rarement vraie (probabilités mal calibrées) ; hypothèse gaussienne.

### Schéma explicatif

**Comment lire le schéma :** chaque classe est modélisée par une gaussienne par variable (courbes de niveau). La frontière est l'endroit où les deux classes ont la même probabilité a posteriori ; elle est lisse et ici courbée car les variances diffèrent.

```python
from sklearn.naive_bayes import GaussianNB
frontiere(GaussianNB(), 'Naive Bayes gaussien : frontière issue de deux lois gaussiennes par classe')
```

### Contextes d'utilisation
- **Cas 1 (gestion)** : filtrage de spam ou classement de tickets de support (avec la variante multinomiale sur des comptes de mots).
- **Cas 2** : diagnostic rapide à partir de peu de mesures cliniques, en première approximation.

- **Métriques** : F1, ROC-AUC ; **calibration** à vérifier car les probabilités sont souvent extrêmes.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Arbre de décision** (*Decision Tree Classifier*, CART) — famille : **classification supervisée**. Objectif : classer par une suite de questions simples sur les variables.
- **Intuition et étapes** : à chaque nœud on cherche la variable et le seuil qui séparent le mieux les classes (groupes les plus « purs »), puis on répète sur chaque branche jusqu'à la profondeur maximale ; la feuille prédit la classe majoritaire.
- **Critère (impureté de Gini)** : $G=1-\sum_{k=1}^{K}p_k^2$ ; on choisit la coupure minimisant la moyenne pondérée de $G$ des deux sous-groupes.
- **Symboles** : $K$ nombre de classes, $p_k$ proportion de la classe $k$ dans le nœud ; $G=0$ pour un nœud pur.
- **Paramètres appris** : variables, seuils, classes des feuilles. **Hyperparamètres** : `max_depth`, `min_samples_leaf`, critère (`gini`/`entropy`).
- **Avantages** : interprétable (règles lisibles), sans standardisation, gère les non-linéarités.
- **Limites** : instable et sujet au surapprentissage si trop profond ; frontières « en rectangles » parallèles aux axes.

### Schéma explicatif

**Comment lire le schéma :** à gauche, l'arbre : chaque nœud affiche sa règle, son impureté (`gini`) et ses effectifs ; les feuilles donnent la classe prédite. À droite, la zone de décision correspondante : des rectangles alignés sur les axes.

```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
arbre = DecisionTreeClassifier(max_depth=3,random_state=GRAINE).fit(X2,y2)
fig,axs = plt.subplots(1,2,figsize=(12,4.2)); plot_tree(arbre,feature_names=['x₁','x₂'],class_names=['0','1'],filled=True,ax=axs[0],fontsize=7)
frontiere(DecisionTreeClassifier(max_depth=3,random_state=GRAINE),'Zone de décision (profondeur 3)',ax=axs[1]); plt.tight_layout(); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (finance/gestion)** : règles d'octroi de crédit lisibles par un conseiller (revenu > seuil, ancienneté > seuil…).
- **Cas 2** : triage médical simplifié ou diagnostic d'une panne par questions successives.
- **Données et prérequis** : tableaux mixtes ; limiter la profondeur ; classes déséquilibrées à traiter (pondération).
- **Peu approprié** : frontière diagonale ou lisse, ou besoin d'un modèle stable : un petit changement de données modifie l'arbre. Préférer une forêt.
- **Métriques** : F1, précision, rappel, ROC-AUC ; profondeur et nombre de feuilles.

## 13. SVM

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Machine à vecteurs de support** (*Support Vector Machine*, SVM) — famille : **classification supervisée**. Objectif : trouver la frontière qui sépare les classes avec la **marge** la plus large possible.
- **Intuition et étapes** : la frontière est déterminée uniquement par les points les plus proches d'elle, les *vecteurs de support*. Avec un noyau (ici RBF), les données sont implicitement projetées dans un espace où une séparation linéaire devient possible.
- **Critère** : $\min_{w,b,\xi}\ \tfrac12\lVert w\rVert^2+C\sum_i\xi_i$ sous $y_i(w^\top\phi(x_i)+b)\ge1-\xi_i$ ; noyau RBF $K(x,x')=\exp(-\gamma\lVert x-x'\rVert^2)$.
- **Symboles** : $w,b$ hyperplan, $\xi_i$ variables d'écart (erreurs tolérées), $\phi$ projection, $C$ pénalité des erreurs, $\gamma$ portée du noyau.
- **Paramètres appris** : vecteurs de support et leurs poids. **Hyperparamètres** : $C$, $\gamma$, type de noyau.
- **Avantages** : efficace en dimension élevée ; frontière non linéaire avec un noyau.
- **Limites** : coût de calcul élevé sur de grands jeux ; pas de probabilités directes et réglage de $C,\gamma$ délicat.

### Schéma explicatif

**Comment lire le schéma :** la ligne pleine est la frontière de décision, les lignes en pointillés les bords de la marge. Les points entourés sont les **vecteurs de support** : ce sont les seuls à influencer la frontière.

```python
from sklearn.svm import SVC
Xs_ = StandardScaler().fit_transform(X2); svm = SVC(kernel='rbf',C=1.0,gamma='scale').fit(Xs_,y2)
gx,gy = np.meshgrid(np.linspace(-3,3,250),np.linspace(-3,3,250)); Z = svm.decision_function(np.c_[gx.ravel(),gy.ravel()]).reshape(gx.shape)
plt.figure(figsize=(5,4)); plt.contourf(gx,gy,Z>0,alpha=.2,cmap='coolwarm'); plt.contour(gx,gy,Z,levels=[-1,0,1],colors='k',linestyles=['--','-','--'])
plt.scatter(Xs_[:,0],Xs_[:,1],c=y2,cmap='coolwarm',edgecolor='k',s=18)
plt.scatter(svm.support_vectors_[:,0],svm.support_vectors_[:,1],s=90,facecolors='none',edgecolors='g',label='vecteurs de support')
plt.title('SVM RBF : frontière, marge et vecteurs de support'); plt.legend(fontsize=8); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (finance)** : classement de documents ou de transactions en dimension élevée avec un échantillon modeste.
- **Cas 2** : reconnaissance de chiffres manuscrits ou classification de séquences biologiques.
- **Données et prérequis** : variables **standardisées** ; taille raisonnable (le coût croît entre $n^2$ et $n^3$) ; réglage de $C$ et $\gamma$ par validation croisée.
- **Peu approprié** : très grands jeux (centaines de milliers de lignes) ou quand on a besoin de probabilités calibrées sans calibration supplémentaire.
- **Métriques** : F1, ROC-AUC (via `decision_function`), précision/rappel.

## 14. Gradient Boosting — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Gradient Boosting pour la classification** (*Gradient Boosting Classifier*, ici `HistGradientBoostingClassifier`) — famille : **classification supervisée** (ensemble séquentiel). Objectif : construire un classifieur précis en empilant des arbres qui corrigent les erreurs des précédents.
- **Intuition et étapes** : on part d'un score constant, puis à chaque itération on ajoute un petit arbre entraîné à réduire la perte (log-loss) en suivant son gradient ; le score final est converti en probabilité par la sigmoïde.
- **Formule** : $F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$, $\;P(y=1\mid x)=\sigma(F_M(x))$.
- **Symboles** : $F_m$ score cumulé après $m$ arbres, $h_m$ arbre ajouté, $\eta$ taux d'apprentissage, $M$ nombre d'itérations.
- **Paramètres appris** : les arbres. **Hyperparamètres** : `max_iter`, `learning_rate`, `max_depth`, `l2_regularization`.
- **Avantages** : excellente performance sur données tabulaires ; gère nativement les valeurs manquantes (version Hist).
- **Limites** : réglage sensible, surapprentissage si trop d'itérations ; moins interprétable.

### Schéma explicatif

**Comment lire le schéma :** à gauche, après 3 itérations seulement, la frontière est grossière ; à droite, après 100 itérations, elle épouse mieux les classes (parfois au prix d'un début de surapprentissage). Le nombre d'itérations est un hyperparamètre clé.

- **Données et prérequis** : données tabulaires de taille moyenne à grande, validation soigneuse (arrêt anticipé), classes déséquilibrées gérées.
- **Peu approprié** : très petits échantillons, images/audio bruts, ou exigence d'explication simple.
- **Métriques** : PR-AUC, ROC-AUC, F1, rappel à précision fixée ; calibration.

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**AdaBoost** (*Adaptive Boosting*) — famille : **classification supervisée** (boosting). Objectif : combiner beaucoup de classifieurs très simples (par défaut des *souches* de profondeur 1) en un classifieur fort.
- **Intuition et étapes** : (1) toutes les observations ont le même poids ; (2) on entraîne un classifieur faible ; (3) on **augmente le poids des observations mal classées** ; (4) on répète, et le vote final est pondéré par la qualité de chaque classifieur.
- **Formules** : $H(x)=\operatorname{sign}\big(\sum_{m=1}^{M}\alpha_m h_m(x)\big)$, avec $\alpha_m=\tfrac12\ln\frac{1-\varepsilon_m}{\varepsilon_m}$ ; mise à jour $w_i\leftarrow w_i\,e^{-\alpha_m y_i h_m(x_i)}$ (cas binaire).
- **Symboles** : $h_m$ classifieur faible, $\varepsilon_m$ son erreur pondérée, $\alpha_m$ son poids dans le vote, $w_i$ poids de l'observation $i$, $y_i\in\{-1,+1\}$.
- **Paramètres appris** : les classifieurs et leurs poids $\alpha_m$. **Hyperparamètres** : `n_estimators`, `learning_rate`, apprenant de base.
- **Avantages** : simple, peu de réglage, bon sur des jeux peu bruités.
- **Limites** : sensible au bruit et aux valeurs aberrantes (il insiste sur les cas difficiles) ; moins performant que les boostings modernes.

### Schéma explicatif

**Comment lire le schéma :** à gauche, un seul classifieur faible (une coupure) ; à droite, 70 souches pondérées : la frontière devient flexible. Les classifieurs successifs se concentrent sur les points que les précédents classaient mal.

```python
from sklearn.ensemble import AdaBoostClassifier
fig,axs = plt.subplots(1,2,figsize=(9,3.5))
for ax,n in zip(axs,(1,70)): frontiere(AdaBoostClassifier(n_estimators=n,random_state=GRAINE),f'AdaBoost : {n} classifieur(s) faible(s)',ax=ax)
plt.tight_layout(); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (marketing)** : classer des prospects à partir de quelques variables binaires ou catégorielles.
- **Cas 2** : détection de visages (algorithme historique de Viola-Jones).
- **Données et prérequis** : jeu peu bruité, étiquettes fiables, classifieurs faibles simples.
- **Peu approprié** : étiquettes bruitées ou nombreuses valeurs aberrantes : l'algorithme sur-pondère justement ces cas.
- **Métriques** : F1, précision, rappel, ROC-AUC.

### Expérience Python fournie (code conservé)

## 16. XGBoost

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**K-moyennes** (*K-means*) — famille : **clustering (non supervisé)**. Objectif : regrouper les observations en $K$ groupes compacts, sans étiquettes.
- **Intuition et étapes** : (1) choisir $K$ centres au hasard ; (2) affecter chaque point au centre le plus proche ; (3) recalculer chaque centre comme la moyenne de son groupe ; répéter (2)-(3) jusqu'à stabilité.
- **Critère (inertie)** : $\min_{C,\mu}\sum_{k=1}^{K}\sum_{x_i\in C_k}\lVert x_i-\mu_k\rVert^2$.
- **Symboles** : $K$ nombre de groupes, $C_k$ groupe $k$, $\mu_k$ son centroïde (moyenne), $x_i$ observation.
- **Paramètres appris** : centroïdes $\mu_k$ et affectations. **Hyperparamètres** : $K$, `n_init` (nombre de relances), initialisation.
- **Avantages** : rapide, simple, passe à l'échelle.
- **Limites** : il faut fixer $K$ ; suppose des groupes sphériques de taille comparable ; sensible aux valeurs aberrantes et à l'initialisation.

### Schéma explicatif

**Comment lire le schéma :** de gauche à droite, trois itérations. Les croix noires sont les centroïdes, la couleur indique le groupe courant. Les centroïdes se déplacent vers le centre de leur groupe jusqu'à se stabiliser.

```python
r = np.random.default_rng(1); C = X_clu[r.choice(len(X_clu),3,replace=False)].copy()
fig,axs = plt.subplots(1,3,figsize=(12,3.4),sharex=True,sharey=True)
for it,ax in enumerate(axs):
    g = ((X_clu[:,None,:]-C[None])**2).sum(2).argmin(1)                  # étape 1 : affectation
    ax.scatter(X_clu[:,0],X_clu[:,1],c=g,cmap='viridis',s=12); ax.scatter(C[:,0],C[:,1],c='k',marker='X',s=140)
    ax.set_title(f'Itération {it+1}',fontsize=10)
    C = np.array([X_clu[g==k].mean(0) if (g==k).any() else C[k] for k in range(3)])   # étape 2 : mise à jour des centroïdes
plt.suptitle('K-means : affectation puis mise à jour des centroïdes (✕)'); plt.tight_layout(); plt.show()
```

- **Données et prérequis** : variables numériques standardisées, groupes de forme approximativement ronde, $K$ choisi (méthode du coude, silhouette).
- **Peu approprié** : groupes de forme allongée ou imbriquée (anneaux, croissants) ou nombreuses valeurs aberrantes.
- **Métriques** : silhouette, inertie, stabilité entre relances, interprétation métier des groupes ; ARI si des étiquettes existent (évaluation externe).

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Classification ascendante hiérarchique** (*Agglomerative Hierarchical Clustering*) — famille : **clustering (non supervisé)**. Objectif : construire une hiérarchie de regroupements, représentée par un dendrogramme.
- **Intuition et étapes** : chaque point est d'abord son propre groupe ; on fusionne à chaque étape les deux groupes les plus proches, jusqu'à ce qu'il n'en reste qu'un ; on « coupe » l'arbre à la hauteur voulue pour obtenir $K$ groupes.
- **Critère de Ward (par défaut)** : on fusionne les groupes $A,B$ qui augmentent le moins la variance intra-groupe : $\Delta(A,B)=\dfrac{|A||B|}{|A|+|B|}\lVert\mu_A-\mu_B\rVert^2$.
- **Symboles** : $A,B$ groupes, $|A|$ effectif, $\mu_A$ centre de $A$, $\Delta$ coût de fusion.
- **Paramètres appris** : l'arbre de fusions. **Hyperparamètres** : `n_clusters` (ou seuil de distance), `linkage` (`ward`, `average`, `single`…), métrique.
- **Avantages** : pas besoin de fixer $K$ à l'avance ; hiérarchie lisible.
- **Limites** : coût de calcul au moins quadratique ; une fusion faite n'est jamais défaite ; sensible aux valeurs aberrantes.

### Schéma explicatif

**Comment lire le schéma :** l'axe horizontal liste les observations, l'axe vertical est la distance de fusion. Plus une branche est haute avant de se rejoindre, plus les deux groupes sont différents. La ligne rouge horizontale montre la coupe qui donne 3 groupes.

```python
from scipy.cluster.hierarchy import linkage, dendrogram
idx = np.random.default_rng(GRAINE).choice(len(X_clu),45,replace=False); Z = linkage(X_clu[idx],method='ward'); coupe = (Z[-3,2]+Z[-2,2])/2     # hauteur qui donne 3 groupes
plt.figure(figsize=(8,3.4)); dendrogram(Z,no_labels=True,color_threshold=coupe); plt.axhline(coupe,c='r',ls='--',label='coupe en 3 groupes')
plt.title('Dendrogramme (45 points tirés au hasard, liaison de Ward)'); plt.ylabel('distance de fusion'); plt.legend(fontsize=8); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (commerce international)** : regrouper des pays selon leur profil d'échanges commerciaux et visualiser la proximité entre pays.
- **Cas 2** : taxonomie de produits ou d'espèces (phylogénie).
- **Données et prérequis** : échantillon de taille modérée (quelques milliers au plus), variables standardisées, choix du lien et de la distance.
- **Peu approprié** : très grands jeux (mémoire quadratique) ou données très bruitées.
- **Métriques** : silhouette, corrélation cophénétique, stabilité, lecture du dendrogramme.

### Expérience Python fournie (code conservé)

```python
from sklearn.cluster import AgglomerativeClustering
etiquettes=AgglomerativeClustering(n_clusters=3).fit_predict(X_clu)
print('Silhouette:',silhouette_score(X_clu,etiquettes))
```

### Analyse de l'expérience Python
- **Entrées** : `X_clu` (240 points). **Hyperparamètres** : `n_clusters=3`, liaison de Ward (défaut).
- **Sortie** : silhouette des étiquettes obtenues.
- **Conclusion possible** : sur des groupes gaussiens bien séparés, K-means et la classification hiérarchique donnent des partitions très proches ; comparer les deux silhouettes.
- **Non conclusif** : la silhouette est calculée avec les mêmes données qui ont servi au regroupement (pas de test possible en non supervisé) ; aucune étiquette de référence n'est utilisée ici.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**DBSCAN** (*Density-Based Spatial Clustering of Applications with Noise*) — famille : **clustering (non supervisé)** et détection de bruit. Objectif : trouver des groupes de forme quelconque comme des zones **denses** séparées par des zones vides.
- **Intuition et étapes** : un point est *central* si au moins `min_samples` points se trouvent à une distance $\varepsilon$. Les points centraux voisins forment un groupe, les points proches d'un centre y sont rattachés, les autres sont du **bruit** (étiquette −1).
- **Définition** : $N_\varepsilon(p)=\{q: d(p,q)\le\varepsilon\}$ ; $p$ est central si $|N_\varepsilon(p)|\ge \text{minPts}$.
- **Symboles** : $\varepsilon$ rayon de voisinage, $\text{minPts}$ effectif minimal, $d$ distance, $N_\varepsilon(p)$ voisinage de $p$.
- **Paramètres appris** : l'appartenance des points (aucun centroïde). **Hyperparamètres** : `eps`, `min_samples`.
- **Avantages** : pas besoin de fixer le nombre de groupes ; formes arbitraires ; détecte les points aberrants.
- **Limites** : sensible au choix de `eps` ; peine quand les densités des groupes diffèrent beaucoup ou en grande dimension.

### Schéma explicatif

**Comment lire le schéma :** sur deux croissants que K-means coupe mal, DBSCAN retrouve les deux formes (couleurs) ; les points noirs ✕ sont classés comme **bruit**.

```python
from sklearn.cluster import DBSCAN, KMeans
Xm,_ = make_moons(300,noise=.07,random_state=GRAINE); lab = DBSCAN(eps=.15,min_samples=5).fit_predict(Xm); km = KMeans(2,n_init=10,random_state=GRAINE).fit_predict(Xm)
fig,axs = plt.subplots(1,2,figsize=(9,3.4),sharex=True,sharey=True)
axs[0].scatter(Xm[:,0],Xm[:,1],c=km,cmap='viridis',s=12); axs[0].set_title('K-means (K=2) : coupe mal les croissants',fontsize=10)
axs[1].scatter(Xm[lab>=0,0],Xm[lab>=0,1],c=lab[lab>=0],cmap='viridis',s=12); axs[1].scatter(Xm[lab<0,0],Xm[lab<0,1],c='k',marker='x',s=30)
axs[1].set_title('DBSCAN : suit la densité, ✕ = bruit',fontsize=10); plt.tight_layout(); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (logistique)** : regrouper des points de livraison géographiques en zones denses et isoler les adresses atypiques.
- **Cas 2** : détection de transactions isolées ou de zones d'activité sur une carte.
- **Données et prérequis** : variables standardisées, distance significative, `eps` choisi avec le graphique des distances au $k$-ième voisin.
- **Peu approprié** : groupes de densités très différentes ou espace de grande dimension (les distances se concentrent).
- **Métriques** : silhouette (hors bruit), proportion de bruit, stabilité, interprétation ; ARI si étiquettes.

### Expérience Python fournie (code conservé)

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

### Analyse de l'expérience Python
- **Entrées** : `X_clu`, non standardisé (les deux axes ont déjà des échelles voisines). **Hyperparamètres** : `eps=0.8`, `min_samples=5`.
- **Sortie** : l'ensemble des étiquettes trouvées (−1 = bruit) et le nombre de points de bruit.
- **Conclusion possible** : le nombre de groupes **n'est pas fixé** : il est déduit de la densité. Si l'ensemble affiché contient −1, ces points sont jugés trop isolés.
- **Non conclusif** : `eps` a été choisi à la main, sans critère ; avec un autre `eps`, le nombre de groupes et de points de bruit change fortement, donc ce résultat n'est pas une mesure objective.

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Mélanges gaussiens** (*Gaussian Mixture Models*, GMM) — famille : **clustering probabiliste (non supervisé)**. Objectif : modéliser les données comme un mélange de lois normales et obtenir une **probabilité d'appartenance** à chaque groupe.
- **Intuition et étapes** : chaque groupe est une gaussienne (centre, forme, poids). L'algorithme **EM** alterne : (E) calculer pour chaque point la probabilité d'appartenance à chaque gaussienne ; (M) réestimer les paramètres avec ces probabilités ; jusqu'à convergence.
- **Modèle** : $p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x\mid\mu_k,\Sigma_k)$, avec $\sum_k\pi_k=1$.
- **Symboles** : $K$ nombre de composantes, $\pi_k$ poids, $\mu_k$ moyenne, $\Sigma_k$ matrice de covariance (forme/orientation du groupe).
- **Paramètres appris** : $\pi_k,\mu_k,\Sigma_k$. **Hyperparamètres** : `n_components`, `covariance_type`, initialisation.
- **Avantages** : groupes elliptiques de tailles différentes, affectation « douce » avec probabilités ; sélection de $K$ par BIC/AIC.
- **Limites** : sensible à l'initialisation (optimum local) ; hypothèse gaussienne ; coût croissant avec la dimension.

### Schéma explicatif

**Comment lire le schéma :** chaque ellipse est une gaussienne du mélange (contour à 2 écarts-types) : sa position est la moyenne, sa forme est la covariance. Un point proche du bord d'une ellipse a une probabilité d'appartenance partagée.

```python
from sklearn.mixture import GaussianMixture
g = GaussianMixture(3,random_state=GRAINE).fit(X_clu); lab = g.predict(X_clu)
fig,ax = plt.subplots(figsize=(5.2,4)); ax.scatter(X_clu[:,0],X_clu[:,1],c=lab,cmap='viridis',s=12)
for mu,cov in zip(g.means_,g.covariances_):
    v,w = np.linalg.eigh(cov); ang = np.degrees(np.arctan2(w[1,1],w[0,1]))
    ax.add_patch(Ellipse(mu,4*np.sqrt(v[1]),4*np.sqrt(v[0]),angle=ang,fill=False,ec='k',lw=2))
ax.set_title('Mélange gaussien : une ellipse (moyenne, covariance) par groupe'); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (marketing)** : segmentation de clients avec probabilité d'appartenance (un client peut être « à 70 % » segment A).
- **Cas 2** : modélisation de densité pour détecter des anomalies, ou séparation de populations en biologie.
- **Données et prérequis** : variables continues, groupes approximativement elliptiques, taille suffisante pour estimer les covariances.
- **Peu approprié** : groupes non gaussiens (croissants), petit échantillon en grande dimension (covariances mal estimées).
- **Métriques** : BIC/AIC, log-vraisemblance, silhouette, stabilité, ARI si étiquettes.

### Expérience Python fournie (code conservé)

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

### Analyse de l'expérience Python
- **Entrées** : `X_clu` ; `GaussianMixture`. **Hyperparamètres** : `n_components=3`, `covariance_type='full'` (défaut), `random_state=42`.
- **Sortie** : les **poids** $\pi_k$ des trois composantes (qui somment à 1) et la silhouette de l'affectation « dure ».
- **Conclusion possible** : des poids proches de 1/3 reflètent les trois groupes de taille égale du générateur (80 points chacun).
- **Non conclusif** : la silhouette ne tient pas compte des probabilités d'appartenance, qui sont la vraie valeur ajoutée du GMM ; pas de comparaison de $K$ par BIC ici.

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Analyse en composantes principales** (*Principal Component Analysis*, ACP/PCA) — famille : **réduction de dimension (non supervisé)**. Objectif : résumer beaucoup de variables corrélées par quelques axes qui conservent le maximum de variance.
- **Intuition et étapes** : on centre et standardise les données, on cherche la direction où les points sont le plus dispersés (1ʳᵉ composante), puis la suivante, orthogonale, etc. On projette ensuite les données sur les $q$ premiers axes.
- **Critère** : $w_1=\arg\max_{\lVert w\rVert=1}\operatorname{Var}(Xw)$ ; les axes sont les vecteurs propres de la matrice de covariance $\Sigma$ ; part de variance expliquée par l'axe $k$ : $\lambda_k/\sum_j\lambda_j$.
- **Symboles** : $X$ données centrées, $w_k$ vecteur de la composante $k$ (*loadings*), $\lambda_k$ valeur propre (variance de l'axe), $q$ nombre de composantes gardées.
- **Paramètres appris** : moyennes et axes $w_k$. **Hyperparamètre** : `n_components`.
- **Avantages** : décorrèle les variables, visualisation 2D, déterministe et rapide.
- **Limites** : uniquement linéaire ; les axes sont des combinaisons de variables difficiles à interpréter ; sensible à l'échelle (standardiser).

### Schéma explicatif

**Comment lire le schéma :** le nuage est allongé selon une diagonale. La flèche rouge (PC1) pointe dans la direction de plus grande dispersion, la flèche verte (PC2) est perpendiculaire et capte ce qui reste. Leur longueur est proportionnelle à l'écart-type porté par chaque axe.

```python
from sklearn.decomposition import PCA
r = np.random.default_rng(GRAINE); A = r.normal(size=(200,2))@np.array([[2.5,1.2],[0,.7]]); pca_s = PCA(2).fit(A); c = A.mean(0)
plt.figure(figsize=(5,4.2)); plt.scatter(A[:,0],A[:,1],s=12,c='gray')
for comp,var,col,nom in zip(pca_s.components_,pca_s.explained_variance_,('r','g'),('PC1','PC2')):
    plt.arrow(c[0],c[1],*(comp*2*np.sqrt(var)),color=col,width=.04,head_width=.25,length_includes_head=True,label=nom)
plt.axis('equal'); plt.title('ACP : axes de variance maximale (données synthétiques)'); plt.legend(); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (finance)** : résumer des dizaines de ratios ou de taux d'intérêt corrélés en 2-3 facteurs (par exemple niveau/pente de la courbe des taux).
- **Cas 2** : prétraitement avant clustering ou régression, ou compression d'images.
- **Données et prérequis** : variables numériques **centrées-réduites**, relations globalement linéaires, pas de valeurs manquantes.
- **Peu approprié** : structure non linéaire (spirales) ou variables catégorielles ; si l'on veut conserver des variables interprétables une à une.
- **Métriques** : variance expliquée (cumulée), éboulis, erreur de reconstruction.

### Expérience Python fournie (code conservé)

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

### Analyse de l'expérience Python
- **Entrées** : jeu **réel** `load_wine` (178 vins, 13 mesures chimiques, UCI), standardisé, projeté sur 2 composantes. **Hyperparamètre** : `n_components=2`.
- **Sortie** : le tableau `explained_variance_ratio_` (part de variance de PC1 et PC2) et un nuage 2D sans couleur.
- **Conclusion possible** : la somme des deux ratios indique quelle part de l'information totale est visible sur le graphique ; le reste est perdu. Le nuage montre s'il existe des regroupements.
- **Non conclusif** : le nuage n'est pas coloré par classe, donc on ne peut pas juger la séparation des cultivars ; la variance expliquée n'est pas la même chose que la pertinence pour une tâche de prédiction.

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**t-SNE** (*t-distributed Stochastic Neighbor Embedding*) — famille : **réduction de dimension non linéaire (non supervisé)**, essentiellement pour la visualisation. Objectif : représenter en 2D des données de grande dimension en **préservant les voisinages**.
- **Intuition et étapes** : on convertit les distances en probabilités de voisinage dans l'espace d'origine ($p_{ij}$), on place les points en 2D avec des probabilités $q_{ij}$ (loi de Student), puis on déplace les points par descente de gradient pour que $q$ ressemble à $p$.
- **Critère** : $\mathrm{KL}(P\Vert Q)=\sum_{i\neq j}p_{ij}\ln\dfrac{p_{ij}}{q_{ij}}$.
- **Symboles** : $p_{ij}$ similarité de $i$ et $j$ en haute dimension (dépend de la *perplexité*), $q_{ij}$ similarité en 2D, KL divergence de Kullback-Leibler.
- **Paramètres appris** : les coordonnées 2D de chaque point. **Hyperparamètres** : `perplexity`, `learning_rate`, nombre d'itérations, graine.
- **Avantages** : révèle très bien des amas locaux ; représentation visuelle parlante.
- **Limites** : les distances entre amas et leurs tailles n'ont pas de sens ; résultat dépendant de la graine et de la perplexité ; pas de transformation de nouveaux points.

### Schéma explicatif

**Comment lire le schéma :** la même donnée (vins) projetée avec deux perplexités. Les couleurs (classes réelles des vins, utilisées **uniquement pour la lecture**, pas pour l'apprentissage) montrent que l'aspect change avec la perplexité : il ne faut pas sur-interpréter la forme exacte.

```python
from sklearn.manifold import TSNE
Xv = StandardScaler().fit_transform(load_wine().data); yv = load_wine().target
fig,axs = plt.subplots(1,2,figsize=(10,3.8))
for ax,p in zip(axs,(5,50)):
    e = TSNE(2,perplexity=p,random_state=GRAINE).fit_transform(Xv); ax.scatter(e[:,0],e[:,1],c=yv,cmap='viridis',s=12); ax.set_title(f't-SNE, perplexité = {p}',fontsize=10)
plt.tight_layout(); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (marketing)** : visualiser des segments de clients ou des plongements de produits pour repérer des familles.
- **Cas 2** : exploration de données génomiques ou d'images pour détecter des amas.
- **Données et prérequis** : jusqu'à quelques dizaines de milliers de points ; variables standardisées ; souvent précédé d'une ACP.
- **Peu approprié** : conclure sur des distances globales, entraîner un modèle de prédiction (pas de fonction de transformation pour de nouveaux points) ou comparer deux graphiques de graines différentes.
- **Métriques** : *trustworthiness* (conservation des voisinages), divergence KL finale ; jamais seulement l'aspect visuel.

### Expérience Python fournie (code conservé)

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

### Analyse de l'expérience Python
- **Entrées** : jeu réel `load_wine` standardisé, 13 variables. **Hyperparamètres** : `perplexity=20`, `random_state=42` (pour reproduire le même dessin).
- **Sortie** : un nuage 2D, sans valeur numérique.
- **Conclusion possible** : s'il apparaît plusieurs amas, il existe une structure locale dans les 13 variables. Le nuage seul ne dit rien du **nombre** de classes ni de leur séparation exacte.
- **Non conclusif** : les axes n'ont pas d'unité ; changer la graine ou la perplexité modifie le dessin. Une évaluation sérieuse exige une mesure (trustworthiness, faite en partie B).

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**UMAP** (*Uniform Manifold Approximation and Projection*) — famille : **réduction de dimension non linéaire (non supervisé)**. Objectif : projeter des données en 2D ou plus en préservant la structure locale **et** mieux la structure globale que t-SNE.
- **Intuition et étapes** : on construit un graphe de voisinage pondéré dans l'espace d'origine (chaque point relié à ses `n_neighbors` plus proches voisins), puis on optimise une disposition en basse dimension dont le graphe ressemble au premier.
- **Critère** : minimisation d'une entropie croisée entre les poids d'arêtes $\mu_{ij}$ (haute dimension) et $\nu_{ij}$ (basse dimension).
- **Symboles** : $\mu_{ij}$ appartenance floue de l'arête $(i,j)$ en haute dimension, $\nu_{ij}$ son équivalent en basse dimension.
- **Paramètres appris** : coordonnées projetées ; le modèle peut ensuite **transformer de nouveaux points**. **Hyperparamètres** : `n_neighbors`, `min_dist`, `metric`, `random_state`.
- **Avantages** : rapide, passe à l'échelle, structure globale mieux conservée, transformation de nouvelles données.
- **Limites** : résultat sensible aux hyperparamètres ; distances entre amas à interpréter avec prudence ; dépendance supplémentaire (`umap-learn`).

### Schéma explicatif

**Comment lire le schéma :** avec peu de voisins (5) UMAP fait ressortir de nombreux petits amas locaux ; avec beaucoup de voisins (50) la vue est plus globale. Les couleurs (classes réelles) ne servent qu'à la lecture.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Forêt d'isolement** (*Isolation Forest*) — famille : **détection d'anomalies (non supervisé)**. Objectif : repérer les observations atypiques sans étiquettes.
- **Intuition et étapes** : on construit des arbres en choisissant à chaque nœud une variable et un seuil **au hasard**. Un point aberrant, isolé, est séparé du reste en peu de coupures ; un point normal, entouré de voisins, en exige beaucoup. La longueur moyenne des chemins donne un score.
- **Score** : $s(x,n)=2^{-E[h(x)]/c(n)}$ avec $c(n)=2H(n-1)-\frac{2(n-1)}{n}$ ; proche de 1 : anomalie, proche de 0,5 ou moins : normal.
- **Symboles** : $h(x)$ longueur du chemin de $x$ dans un arbre, $E[h(x)]$ sa moyenne sur la forêt, $n$ taille d'échantillon, $c(n)$ longueur moyenne normalisante, $H$ nombre harmonique.
- **Paramètres appris** : les arbres d'isolement. **Hyperparamètres** : `n_estimators`, `contamination` (part attendue d'anomalies), `max_samples`.
- **Avantages** : rapide, sans hypothèse de distribution, adapté aux données volumineuses.
- **Limites** : `contamination` fixe arbitrairement le nombre d'alertes ; performances moindres sur anomalies « noyées » dans la masse ou en très grande dimension.

### Schéma explicatif

**Comment lire le schéma :** la couleur donne le score d'anomalie (plus foncé = plus atypique). Les points normaux sont au cœur des groupes ; les deux points isolés aux extrémités (croix rouges) sont ceux qui seraient isolés le plus rapidement.

```python
from sklearn.ensemble import IsolationForest
Xa = np.vstack([X_clu,np.array([[9,9],[-8,-8]])]); iso = IsolationForest(random_state=GRAINE).fit(Xa)
gx,gy = np.meshgrid(np.linspace(-11,12,200),np.linspace(-11,12,200)); S = iso.score_samples(np.c_[gx.ravel(),gy.ravel()]).reshape(gx.shape)
plt.figure(figsize=(5.4,4.2)); plt.contourf(gx,gy,S,levels=15,cmap='viridis'); plt.colorbar(label='score (plus bas = plus anormal)')
plt.scatter(Xa[:,0],Xa[:,1],c='w',s=8); plt.scatter(Xa[-2:,0],Xa[-2:,1],c='r',marker='x',s=80,label='anomalies injectées')
plt.title("Forêt d'isolement : carte des scores"); plt.legend(fontsize=8); plt.show()
```

- **Cas 1 (finance)** : repérer des transactions bancaires inhabituelles pour examen par un analyste (fraude).
- **Cas 2** : surveiller des capteurs industriels ou des journaux de connexion.
- **Données et prérequis** : variables numériques, anomalies rares et différentes de la masse ; décision sur le niveau d'alerte acceptable.
- **Peu approprié** : quand les anomalies sont nombreuses ou forment un amas dense (elles ne sont plus « isolées »), ou pour des séries temporelles sans prétraitement.
- **Métriques** : qualité des alertes (examen manuel), précision/rappel, PR-AUC **uniquement si des étiquettes fiables existent**.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Bandits manchots** (*Multi-armed Bandits*), exploration uniforme — famille : **apprentissage par renforcement** (cas à un seul état). Objectif : maximiser la récompense cumulée en choisissant entre plusieurs actions (« bras ») dont les gains sont inconnus.
- **Intuition et étapes** : à chaque tour l'agent choisit un bras, observe une récompense (0 ou 1) et met à jour son estimation de la valeur de ce bras. Ici la stratégie est la plus simple : on **alterne les bras à tour de rôle** (exploration pure, aucune exploitation).
- **Estimation et critère** : $\hat Q_t(a)=\dfrac1{N_t(a)}\sum_{s\le t:\,a_s=a}r_s$ ; objectif : minimiser le **regret** $R_T=T\mu^\*-\sum_{t=1}^{T}r_t$.
- **Symboles** : $a$ bras, $N_t(a)$ nombre de tirages du bras $a$ jusqu'à $t$, $r_s$ récompense, $\mu^\*$ espérance du meilleur bras, $T$ horizon.
- **Paramètres appris** : valeurs estimées $\hat Q(a)$. **Hyperparamètres** : horizon $T$ ; ici aucun paramètre de stratégie.
- **Avantages** : estimations non biaisées et équilibrées de tous les bras ; très simple.
- **Limites** : gaspille des essais sur les mauvais bras (regret qui croît linéairement) ; ne s'adapte pas à ce qu'il apprend.

### Schéma explicatif

**Comment lire le schéma :** boucle agent–environnement. L'agent choisit un bras, l'environnement renvoie une récompense, l'agent met à jour sa moyenne estimée pour ce bras puis recommence. Ici le choix du bras est fixé par rotation, indépendamment des gains observés.

```python
flux(['Agent\nchoisit un bras (rotation)','Environnement\nrenvoie 0 ou 1','Agent met à jour\nla moyenne du bras','Tour suivant'],
     'Boucle agent–environnement du bandit à exploration uniforme')
```

### Contextes d'utilisation
- **Cas 1 (marketing)** : tester trois versions d'une bannière publicitaire en les affichant à tour de rôle (test A/B/C classique).
- **Cas 2** : comparer plusieurs traitements en essai clinique avec une allocation équilibrée.
- **Données et prérequis** : possibilité de répéter l'action et d'observer une récompense rapidement ; récompenses indépendantes et stables dans le temps.
- **Peu approprié** : quand chaque essai a un coût élevé (une mauvaise option est coûteuse) : on préfère une stratégie qui exploite vite le meilleur bras.
- **Métriques** : récompense cumulée, regret cumulé, taux de choix du meilleur bras, variation entre graines.

### Expérience Python fournie (code conservé)

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3)
for t in range(500):
    bras=t%3
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Valeurs estimées:',somme/comptes)
```

### Analyse de l'expérience Python
- **Entrées** : trois bras de Bernoulli de probabilités de succès *vraies* 0,2 / 0,5 / 0,7 (**simulation**, pas un jeu de données réel), 500 tours, graine 42. **Hyperparamètres** : aucun de stratégie ; le tirage se fait en rotation (`t%3`).
- **Sortie** : les valeurs estimées de chaque bras (moyennes empiriques).
- **Conclusion possible** : chaque bras est tiré environ 167 fois ; l'écart-type d'une moyenne de proportion est de l'ordre de 0,04, donc les estimations doivent être proches (à quelques centièmes) des vraies probabilités ; l'ordre 0,2 < 0,5 < 0,7 est retrouvé.
- **Non conclusif** : cette stratégie n'optimise pas la récompense (un tiers des essais va au pire bras) ; la comparaison avec une stratégie adaptative est l'objet de l'exemple suivant. Aucune conclusion sur un problème économique réel.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Epsilon-greedy** (*ε-greedy*) — famille : **apprentissage par renforcement** (bandits). Objectif : arbitrer entre **exploration** (essayer) et **exploitation** (jouer le meilleur bras connu).
- **Intuition et étapes** : avec probabilité $\varepsilon$ on choisit un bras au hasard (explorer) ; sinon on choisit le bras de meilleure valeur estimée (exploiter). On observe la récompense et on met à jour l'estimation.
- **Règle** : $a_t=\begin{cases}\text{bras aléatoire}&\text{avec probabilité }\varepsilon\\\arg\max_a\hat Q_t(a)&\text{avec probabilité }1-\varepsilon\end{cases}$
- **Symboles** : $\varepsilon\in[0,1]$ taux d'exploration, $\hat Q_t(a)$ valeur estimée du bras $a$, $a_t$ bras choisi.
- **Paramètres appris** : $\hat Q(a)$. **Hyperparamètres** : $\varepsilon$ (constant ou décroissant), initialisation des valeurs.
- **Avantages** : très simple, adaptatif, réduit fortement le regret par rapport à l'exploration pure.
- **Limites** : $\varepsilon$ fixe continue d'explorer inutilement (regret linéaire) ; l'exploration est aveugle (pas d'incertitude), contrairement à UCB ou Thompson sampling.

### Schéma explicatif

**Comment lire le schéma :** part de tours où le meilleur bras est choisi, en fonction du temps (moyenne sur 30 répétitions). Avec $\varepsilon=0$ l'agent peut se bloquer sur un mauvais bras ; avec $\varepsilon=0{,}1$ il converge vers le meilleur bras ; avec $\varepsilon=0{,}5$ il explore trop et plafonne.

```python
def eps_greedy(eps,T=500,graine=0,p=np.array([.2,.5,.7])):
    g=np.random.default_rng(graine); s=np.zeros(3); n=np.zeros(3); best=np.zeros(T)
    for t in range(T):
        a = int(g.integers(3)) if g.random()<eps else int(np.argmax(s/np.maximum(n,1)))
        s[a]+= g.random()<p[a]; n[a]+=1; best[t]=(a==2)
    return best
plt.figure(figsize=(6,3.5))
for eps in (0,.1,.5):
    m = np.mean([eps_greedy(eps,graine=k) for k in range(30)],axis=0); plt.plot(np.convolve(m,np.ones(20)/20,mode='valid'),label=f'ε = {eps}')
plt.title('ε-greedy : proportion de choix du meilleur bras'); plt.xlabel('tour'); plt.ylabel('P(choisir le meilleur bras)'); plt.legend(); plt.show()
```

### Contextes d'utilisation
- **Cas 1 (marketing/e-commerce)** : choisir dynamiquement quelle offre promotionnelle afficher pour maximiser le taux de clic.
- **Cas 2** : choix d'un algorithme de recommandation ou d'un itinéraire de livraison avec apprentissage en ligne.
- **Données et prérequis** : retour d'information immédiat après chaque action ; comportement des bras stable ou lentement variable.
- **Peu approprié** : environnements non stationnaires rapides sans oubli (les estimations deviennent fausses) ou quand une exploration aléatoire est inacceptable (risque, sécurité).
- **Métriques** : récompense cumulée, regret, proportion du meilleur bras, variation entre graines.

### Expérience Python fournie (code conservé)

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Analyse de l'expérience Python
- **Entrées** : mêmes trois bras (0,2 / 0,5 / 0,7), **simulation**, 1000 tours. **Hyperparamètre** : `epsilon=0.1`.
- **Sortie** : le nombre de tirages par bras et les valeurs estimées.
- **Point d'attention dans le code** : la valeur utilisée pour choisir est `(somme+1)/(comptes+1)`. Cela revient à ajouter un succès fictif à chaque bras (initialisation **optimiste**, valeur 1 avant tout tirage), ce qui force chaque bras à être essayé au moins une fois. Les valeurs *affichées*, elles, utilisent `somme/max(comptes,1)` (sans succès fictif).
- **Conclusion possible** : le bras 2 (0,7) devrait concentrer l'essentiel des tirages ; les bras 0 et 1 reçoivent surtout des tirages d'exploration. Comme ils sont peu tirés, leurs estimations sont moins précises que celle du meilleur bras.
- **Non conclusif** : une seule graine ; pour juger l'algorithme il faut moyenner sur plusieurs graines (fait en partie B).

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Q-learning** — famille : **apprentissage par renforcement** (sans modèle, *off-policy*, tabulaire). Objectif : apprendre une politique qui maximise la récompense cumulée dans un environnement à états et actions discrets.
- **Intuition et étapes** : l'agent maintient une table $Q(s,a)$ = récompense future espérée si l'on fait $a$ dans l'état $s$ puis que l'on agit au mieux. À chaque pas il observe $(s,a,r,s')$ et rapproche $Q(s,a)$ d'une cible construite avec la **meilleure** action suivante.
- **Mise à jour** : $Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\big]$.
- **Symboles** : $s,s'$ états courant/suivant, $a$ action, $r$ récompense, $\alpha$ taux d'apprentissage, $\gamma$ facteur d'actualisation, $\max_{a'}Q(s',a')$ meilleure valeur à l'état suivant.
- **Paramètres appris** : la table $Q$. **Hyperparamètres** : $\alpha$ (0,3 ici), $\gamma$ (0,95), $\varepsilon$ d'exploration (de 1 à 0,05), nombre d'épisodes.
- **Avantages** : simple, converge vers la politique optimale en environnement tabulaire avec exploration suffisante ; apprend sur la politique d'exploration tout en visant la politique gloutonne.
- **Limites** : la table explose avec le nombre d'états ; instable ou optimiste (biais de maximisation) ; peut longer des zones risquées pendant l'entraînement.

### Schéma explicatif

**Comment lire le schéma :** l'agent observe l'état, choisit une action (ε-greedy), l'environnement renvoie récompense et nouvel état, puis la table $Q$ est mise à jour avec la **meilleure** valeur du nouvel état. La boucle recommence jusqu'à la fin de l'épisode.

```python
flux(['État s','Action a\n(ε-greedy sur Q)','Environnement\n(r, s′)','Mise à jour\nQ(s,a) += α[r+γ max Q(s′,·)−Q(s,a)]','s ← s′'],
     'Boucle du Q-learning (la cible utilise le max)')
```

### Contextes d'utilisation
- **Cas 1 (logistique/gestion)** : politique de réapprovisionnement ou de routage simplifié avec peu d'états discrets (niveau de stock, jour).
- **Cas 2** : jeux simples et petits problèmes de navigation (labyrinthes).
- **Données et prérequis** : environnement simulable ou accessible par interaction, états et actions discrets en petit nombre, fonction de récompense bien conçue.
- **Peu approprié** : états continus ou très nombreux (images, finance haute fréquence) : utiliser une approximation de fonction (DQN) ; ou quand une exploration réelle coûte cher.
- **Métriques** : récompense cumulée par épisode, taux de réussite, stabilité entre graines, vitesse de convergence.

### Expérience Python fournie (code conservé)

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python
- **Entrées** : environnement **Gymnasium** `FrozenLake-v1` sans glissement (`is_slippery=False`) : grille 4×4, 16 états, 4 actions (0 gauche, 1 bas, 2 droite, 3 haut). L'environnement est une **simulation déterministe**, pas un jeu de données.
- **Hyperparamètres** : 4000 épisodes, $\alpha=0.3$, $\gamma=0.95$, ε décroissant de 1 à 0,05, graine 42.
- **Sortie** : `Q.argmax(axis=1)`, l'action gloutonne par état (une grille de 16 chiffres).
- **Conclusion possible** : on lit la politique apprise sur le chemin menant de l'état 0 au but (état 15). **Attention** : les états-trou et l'état but gardent une ligne $Q$ nulle, donc leur `argmax` vaut 0 sans signification ; seuls les états du chemin sont informatifs.
- **Non conclusif** : la sortie ne donne ni récompense cumulée ni taux de réussite ; il faudrait exécuter la politique pour la valider (fait en partie B sur plusieurs graines).

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**SARSA** (*State–Action–Reward–State–Action*) — famille : **apprentissage par renforcement** (sans modèle, *on-policy*, tabulaire). Objectif : apprendre la valeur d'une politique **telle qu'elle est effectivement suivie**, exploration comprise.
- **Intuition et étapes** : identique au Q-learning, sauf que la cible utilise l'action $a'$ **réellement choisie** à l'état suivant (par la politique ε-greedy) et non la meilleure action possible. SARSA est donc « prudent » : il tient compte de ses propres erreurs d'exploration.
- **Mise à jour** : $Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\,Q(s',a')-Q(s,a)\big]$.
- **Symboles** : $(s,a,r,s',a')$ le quintuplet qui donne son nom à l'algorithme, $\alpha$ taux d'apprentissage, $\gamma$ facteur d'actualisation.
- **Paramètres appris** : la table $Q$. **Hyperparamètres** : $\alpha$, $\gamma$, schéma de $\varepsilon$, nombre d'épisodes.
- **Avantages** : politique plus sûre près des états dangereux pendant l'apprentissage ; plus stable.
- **Limites** : apprend la politique d'exploration (sous-optimale tant que $\varepsilon>0$) ; convergence plus lente vers la politique gloutonne optimale.

### Schéma explicatif

**Comment lire le schéma :** la différence avec le Q-learning est dans la boîte de mise à jour. Ici l'action suivante $a'$ est choisie **avant** la mise à jour et utilisée dans la cible : la boucle comporte une étape de plus (choix de $a'$).

```python
flux(['État s, action a','Environnement\n(r, s′)','Choisir a′\n(ε-greedy sur Q)','Mise à jour\nQ(s,a) += α[r+γ Q(s′,a′)−Q(s,a)]','s←s′, a←a′'],
     'Boucle de SARSA (la cible utilise l’action réellement choisie a′)')
```

### Contextes d'utilisation
- **Cas 1 (logistique)** : apprentissage en ligne d'une politique d'acheminement où les erreurs d'exploration ont un coût réel (retards, pénalités).
- **Cas 2** : robot qui évite des zones dangereuses (falaise) pendant son apprentissage.
- **Données et prérequis** : mêmes que le Q-learning (états/actions discrets, environnement interactif) ; schéma de décroissance de ε défini.
- **Peu approprié** : quand on souhaite apprendre directement la politique gloutonne optimale à partir de données collectées par une autre politique (préférer le Q-learning *off-policy*).
- **Métriques** : récompense cumulée **pendant l'apprentissage** (où SARSA est souvent meilleur) et en évaluation gloutonne, taux de réussite, variation entre graines.

### Expérience Python fournie (code conservé)

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python
- **Entrées** : même environnement déterministe `FrozenLake-v1` (`is_slippery=False`), **simulation**. **Hyperparamètres** : mêmes que l'exemple 27 (4000 épisodes, $\alpha=0.3$, $\gamma=0.95$, ε de 1 à 0,05).
- **Sortie** : l'action gloutonne par état (`Politique SARSA`).
- **Conclusion possible** : sur cette grille sans risque stochastique, SARSA et Q-learning aboutissent en général à des politiques équivalentes sur le chemin utile ; **leur différence théorique n'est pas visible ici** car le chemin optimal est unique et l'environnement est déterministe.
- **Non conclusif** : pour voir la différence entre *on-policy* et *off-policy* il faut un environnement où l'exploration est risquée (par exemple `CliffWalking-v1`), ce que fait la partie B.

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
**Deep Q-Network** (DQN) — famille : **apprentissage par renforcement profond** (valeur, *off-policy*). Objectif : étendre le Q-learning aux espaces d'états grands ou continus en remplaçant la table $Q$ par un **réseau de neurones**.
- **Intuition et étapes** : un réseau $Q(s,a;\theta)$ prédit la valeur de chaque action. Pour stabiliser l'apprentissage, on utilise (1) une **mémoire de rejeu** : on réentraîne sur des transitions tirées au hasard dans le passé ; (2) un **réseau cible** $\theta^-$ copié périodiquement, qui fournit la cible de Bellman.
- **Cible et perte** : $y=r+\gamma(1-d)\max_{a'}Q(s',a';\theta^-)$ ; $\;L(\theta)=\mathbb E\big[(y-Q(s,a;\theta))^2\big]$.
- **Symboles** : $\theta$ poids du réseau en ligne, $\theta^-$ poids du réseau cible, $d\in\{0,1\}$ indicateur de fin d'épisode, $\gamma$ facteur d'actualisation, $\mathbb E$ espérance sur un mini-lot.
- **Paramètres appris** : $\theta$. **Hyperparamètres** : taille de la mémoire, taille du lot, taux d'apprentissage, fréquence de mise à jour de $\theta^-$, $\varepsilon$.
- **Avantages** : traite des états continus ou de grande dimension (images) ; généralise entre états voisins.
- **Limites** : instable et gourmand en échantillons ; nombreux hyperparamètres, résultats très variables selon la graine.

### Schéma explicatif

**Comment lire le schéma :** l'agent agit dans l'environnement ; chaque transition est stockée dans la mémoire de rejeu. Des mini-lots tirés au hasard servent à calculer la cible avec le réseau cible, puis à mettre à jour le réseau en ligne. Le réseau cible n'est recopié que de temps en temps.

```python
flux(['Environnement\n(s, r, s′)','Réseau en ligne\nQ(s,a;θ)\nchoisit a','Mémoire de rejeu\n(s,a,r,s′,d)','Mini-lot aléatoire\n→ cible y','Réseau cible θ⁻\n(copie périodique)'],
     'Architecture du DQN : réseau en ligne, mémoire de rejeu, réseau cible', largeur_boite=15)
```

### Contextes d'utilisation
- **Cas 1 (gestion/finance)** : gestion dynamique de stocks ou d'exécution d'ordres où l'état est un vecteur continu (prix, volumes, stocks).
- **Cas 2** : jeux vidéo (Atari) et contrôle simple (CartPole).
- **Données et prérequis** : simulateur rapide, beaucoup d'interactions, matériel adapté (GPU utile), fonction de récompense soigneusement conçue.
- **Peu approprié** : petits problèmes tabulaires (le Q-learning suffit) ou environnements où l'on ne peut pas simuler ni expérimenter librement.
- **Métriques** : récompense cumulée moyenne sur plusieurs graines, taux de réussite, courbe d'apprentissage, variation entre graines.

## Q7–Q8 — Régression (algorithmes 1 à 7)

**Référence** : `DummyRegressor` (moyenne) et `LinearRegression`. **Méthodes avancées** : forêt aléatoire et gradient boosting. Les hyperparamètres sont choisis sur la **validation** (MAE minimale) ; le test n'est lu qu'à la fin. Le temps affiché inclut la recherche d'hyperparamètres (coût de calcul).

```python
# ── Outils communs de sélection sur validation ────────────────────────────────
from itertools import product
from sklearn.base import clone
from sklearn.pipeline import Pipeline
from sklearn.dummy import DummyRegressor, DummyClassifier
from sklearn.metrics import mean_squared_error, precision_score, recall_score, roc_auc_score, average_precision_score

def choisir_sur_validation(modele, grille, Xtr, ytr, Xva, yva, score):
    """Ajuste le modèle pour chaque combinaison de la grille sur l'entraînement ; garde celle qui maximise `score` sur la validation."""
    t0 = time.time(); meilleur = None
    combinaisons = list(product(*grille.values())) if grille else [()]
    for valeurs in combinaisons:
        params = dict(zip(grille.keys(), valeurs))
        m = clone(modele).set_params(**params).fit(Xtr, ytr)
        s = score(m, Xva, yva)
        if meilleur is None or s > meilleur[0]: meilleur = (s, params, m)
    return meilleur[2], meilleur[1], time.time()-t0

def metriques_reg(y, p): return {'MAE': mean_absolute_error(y,p), 'RMSE': float(np.sqrt(mean_squared_error(y,p))), 'R²': r2_score(y,p)}
score_reg = lambda m, X, y: -mean_absolute_error(y, m.predict(X))      # on maximise −MAE
```

```python
std = lambda: StandardScaler()
MODELES_REG = {
  '0. Référence naïve (moyenne)': (DummyRegressor(), {}),
  '1. Régression linéaire':        (Pipeline([('std',std()),('m',LinearRegression())]), {}),
  '2. Ridge':                      (Pipeline([('std',std()),('m',Ridge())]), {'m__alpha':[.1,1,10,100]}),
  '3. Lasso':                      (Pipeline([('std',std()),('m',Lasso(max_iter=20000))]), {'m__alpha':[1e-4,1e-3,1e-2,.1]}),
  '4. Polynomiale (Ridge)':        (Pipeline([('std',std()),('poly',PolynomialFeatures(include_bias=False)),('std2',std()),('m',Ridge())]), {'poly__degree':[2,3],'m__alpha':[1,10,100]}),
  '5. Arbre de régression':        (DecisionTreeRegressor(random_state=GRAINE), {'max_depth':[4,8,12,16],'min_samples_leaf':[5,20]}),
  '6. Forêt aléatoire':            (RandomForestRegressor(n_estimators=100,random_state=GRAINE,n_jobs=-1), {'max_depth':[12,None],'min_samples_leaf':[1,5]}),
  '7. Gradient boosting (Hist)':   (HistGradientBoostingRegressor(random_state=GRAINE), {'learning_rate':[.05,.1],'max_iter':[200,500]}),
}
lignes, MEILLEURS_REG = [], {}
for nom,(m,g) in MODELES_REG.items():
    mod, params, duree = choisir_sur_validation(m, g, X1_tr, y1_tr, X1_va, y1_va, score_reg)
    MEILLEURS_REG[nom] = mod
    lignes.append({'Modèle':nom, **{k:round(v,4) for k,v in metriques_reg(y1_te, mod.predict(X1_te)).items()}, 'MAE validation':round(-score_reg(mod,X1_va,y1_va),4), 'Temps (s)':round(duree,1), 'Hyperparamètres retenus':params})
TAB_REG = pd.DataFrame(lignes).set_index('Modèle'); RESULTATS['regression'] = TAB_REG
display(TAB_REG.sort_values('MAE'))
```

```python
def score_clf(m, X, y): return f1_score(y, m.predict(X))
def score_de(m, X):  return m.predict_proba(X)[:,1] if hasattr(m,'predict_proba') else m.decision_function(X)
def metriques_clf(m, X, y):
    p = m.predict(X); s = score_de(m, X)
    return {'Précision':precision_score(y,p,zero_division=0),'Rappel':recall_score(y,p),'F1':f1_score(y,p),'ROC-AUC':roc_auc_score(y,s),'PR-AUC':average_precision_score(y,s)}
def pipe(est): return Pipeline([('prep',prep()),('m',est)])
MODELES_CLF = {
  '0. Référence naïve (majoritaire)': (pipe(DummyClassifier(strategy='most_frequent')), {}),
  '8. Régression logistique':   (pipe(LogisticRegression(max_iter=2000,random_state=GRAINE)), {'m__C':[.01,.1,1,10]}),
  '9. KNN':                     (pipe(KNeighborsClassifier()), {'m__n_neighbors':[5,15,31]}),
  '10. Naive Bayes':            (pipe(GaussianNB()), {}),
  '11. Arbre de décision':      (pipe(DecisionTreeClassifier(random_state=GRAINE)), {'m__max_depth':[3,5,8],'m__min_samples_leaf':[5,30]}),
  '12. Forêt aléatoire':        (pipe(RandomForestClassifier(n_estimators=150,random_state=GRAINE,n_jobs=-1)), {'m__max_depth':[8,None],'m__min_samples_leaf':[1,5]}),
  '13. SVM (RBF)':              (pipe(SVC(kernel='rbf',random_state=GRAINE)), {'m__C':[1,10]}),
  '14. Gradient boosting (Hist)':(pipe(HistGradientBoostingClassifier(random_state=GRAINE)), {'m__learning_rate':[.05,.1],'m__max_iter':[100,200]}),
  '15. AdaBoost':               (pipe(AdaBoostClassifier(random_state=GRAINE)), {'m__n_estimators':[50,200],'m__learning_rate':[.5,1.]}),
}
try:
    from xgboost import XGBClassifier
    MODELES_CLF['16. XGBoost'] = (pipe(XGBClassifier(random_state=GRAINE,eval_metric='logloss')), {'m__max_depth':[3,5],'m__n_estimators':[100,300],'m__learning_rate':[.05,.1]})
except ImportError:
    print('XGBoost absent : ligne 16 omise.')
lignes, MEILLEURS_CLF = [], {}
for nom,(m,g) in MODELES_CLF.items():
    mod, params, duree = choisir_sur_validation(m, g, X2_tr, y2_tr, X2_va, y2_va, score_clf)
    MEILLEURS_CLF[nom] = mod
    lignes.append({'Modèle':nom, **{k:round(v,4) for k,v in metriques_clf(mod,X2_te,y2_te).items()}, 'F1 validation':round(score_clf(mod,X2_va,y2_va),4), 'Temps (s)':round(duree,1), 'Hyperparamètres retenus':params})
TAB_CLF = pd.DataFrame(lignes).set_index('Modèle'); RESULTATS['classification'] = TAB_CLF
display(TAB_CLF.sort_values('PR-AUC',ascending=False))
```

## Q6–Q8 — Clustering et réduction de dimension sur D3 (Wine, algorithmes 17 à 23)

Les **étiquettes réelles (cultivars) ne sont jamais utilisées pour apprendre** : elles servent uniquement à l'évaluation externe (ARI, NMI) et à la colorisation des figures. La standardisation est apprise sur les 178 vins (aucune étiquette n'intervient, et il n'y a pas de jeu de test en non supervisé).

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import adjusted_rand_score, normalized_mutual_info_score
vin = load_wine(as_frame=True); Xv = StandardScaler().fit_transform(vin.data); yv = vin.target.values
print('Wine :', Xv.shape, '| valeurs manquantes :', int(vin.data.isna().sum().sum()), '| répartition des classes :', np.bincount(yv))
# Choix de K par silhouette pour K-means (critère interne)
sil_k = {k: silhouette_score(Xv, KMeans(k,n_init=10,random_state=GRAINE).fit_predict(Xv)) for k in range(2,9)}
print('Silhouette K-means selon K :', {k:round(v,3) for k,v in sil_k.items()})
# DBSCAN : eps choisi par silhouette interne (hors bruit), min_samples = 5
meilleur_eps = None
for eps in np.arange(1.2,3.6,.2):
    l = DBSCAN(eps=eps,min_samples=5).fit_predict(Xv); ok = l>=0
    if len(set(l[ok]))>=2 and ok.sum()>10 and (l==-1).mean()<=.20:     # on refuse les eps qui rejettent plus de 20 % des points
        s = silhouette_score(Xv[ok],l[ok])
        if meilleur_eps is None or s>meilleur_eps[1]: meilleur_eps=(round(float(eps),1),s)
if meilleur_eps is None: print('Aucun eps ne donne ≥ 2 groupes avec ≤ 20 % de bruit : DBSCAN ne trouve pas de structure multi-groupes ici (eps=3,0 par défaut).')
EPS = meilleur_eps[0] if meilleur_eps else 3.0; print('eps utilisé pour DBSCAN :', EPS)
```

```python
MODELES_CLU = {
  '17. K-means (K=3)':               KMeans(3,n_init=10,random_state=GRAINE),
  '18. Hiérarchique (Ward, K=3)':    AgglomerativeClustering(3,linkage='ward'),
  '19. DBSCAN':                      DBSCAN(eps=EPS,min_samples=5),
  '20. Mélange gaussien (K=3)':      GaussianMixture(3,random_state=GRAINE),
}
def silhouette_sans_bruit(X,l):
    ok = l>=0
    return silhouette_score(X[ok],l[ok]) if len(set(l[ok]))>=2 else np.nan
def stabilite(modele, X, ref, n=20):
    """ARI moyen entre la partition de référence (toutes les données) et la partition obtenue sur 80 % des points."""
    r = np.random.default_rng(GRAINE); a=[]
    for _ in range(n):
        idx = r.choice(len(X), int(.8*len(X)), replace=False); a.append(adjusted_rand_score(ref[idx], clone(modele).fit_predict(X[idx])))
    return float(np.mean(a)), float(np.std(a))
lignes = []
for nom,m in MODELES_CLU.items():
    t0=time.time(); l = clone(m).fit_predict(Xv); dt=time.time()-t0; st = stabilite(m,Xv,l)
    lignes.append({'Méthode':nom,'Nb groupes':len(set(l))-(1 if -1 in l else 0),'% bruit':round(100*(l==-1).mean(),1),'Silhouette (hors bruit)':round(silhouette_sans_bruit(Xv,l),3),
                   'ARI (externe)':round(adjusted_rand_score(yv,l),3),'NMI (externe)':round(normalized_mutual_info_score(yv,l),3),'Stabilité ARI (moy)':round(st[0],3),'Stabilité ARI (écart-type)':round(st[1],3),'Temps (s)':round(dt,3)})
TAB_CLU = pd.DataFrame(lignes).set_index('Méthode'); RESULTATS['clustering'] = TAB_CLU; display(TAB_CLU)
```

**Interprétation (à adapter à vos chiffres).** Les **critères internes** (silhouette, stabilité) jugent la cohésion des groupes sans connaître la vérité ; les **critères externes** (ARI, NMI) mesurent l'accord avec les 3 cultivars réels (1 = accord parfait, 0 ≈ hasard). Un bon score interne n'implique pas un bon score externe : par exemple, la silhouette maximale selon $K$ peut désigner 2 groupes alors que la réalité en compte 3. Référence : K-means ; méthodes avancées : mélange gaussien (formes elliptiques) et DBSCAN (densité). DBSCAN (eps choisi par silhouette interne, avec au plus 20 % de bruit) est ici handicapé par la dimension 13 (les distances se concentrent) et la petite taille de l'échantillon, ce qui se voit dans la part de bruit et dans le nombre de groupes obtenu. **Limites** : 178 observations seulement ; les groupes n'ont de sens métier que s'ils sont interprétés (profils chimiques moyens par groupe).

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE, trustworthiness
t0=time.time(); pca = PCA().fit(Xv); P_pca = pca.transform(Xv)[:,:2]; t_pca=time.time()-t0
cum = np.cumsum(pca.explained_variance_ratio_)
print('Variance expliquée — PC1:',round(pca.explained_variance_ratio_[0],3),'| PC2:',round(pca.explained_variance_ratio_[1],3),'| cumulée à 2 :',round(cum[1],3),'| composantes pour 90 % :',int(np.argmax(cum>=.9))+1)
t0=time.time(); P_tsne = TSNE(2,perplexity=20,random_state=GRAINE).fit_transform(Xv); t_tsne=time.time()-t0
projections = {'21. ACP (2 composantes)':(P_pca,t_pca), '22. t-SNE (perplexité 20)':(P_tsne,t_tsne)}
try:
    import umap
    t0=time.time(); P_umap = umap.UMAP(n_neighbors=15,random_state=GRAINE).fit_transform(Xv); projections['23. UMAP (15 voisins)']=(P_umap,time.time()-t0)
except ImportError: print('UMAP absent : ligne 23 omise.')
lignes=[{'Méthode':n,'Trustworthiness (10 voisins)':round(trustworthiness(Xv,P,n_neighbors=10),3),'Temps (s)':round(t,2)} for n,(P,t) in projections.items()]
TAB_DIM = pd.DataFrame(lignes).set_index('Méthode'); RESULTATS['reduction'] = TAB_DIM; display(TAB_DIM)
fig,axs = plt.subplots(1,len(projections),figsize=(4.4*len(projections),3.8))
for ax,(n,(P,_)) in zip(np.atleast_1d(axs),projections.items()):
    ax.scatter(P[:,0],P[:,1],c=yv,cmap='viridis',s=14); ax.set_title(n,fontsize=10)
plt.tight_layout(); plt.show()
```

**Figure 3.** Projections 2D des 178 vins ; la couleur est le cultivar réel (utilisé uniquement pour la lecture). **Interprétation.** La *trustworthiness* (de 0 à 1) mesure si les voisins d'un point dans la projection étaient déjà ses voisins dans l'espace d'origine : plus elle est proche de 1, mieux les voisinages sont conservés. Référence : ACP (linéaire, rapide, **interprétable** par ses axes et sa variance expliquée) ; méthodes avancées : t-SNE et UMAP, qui préservent mieux les voisinages locaux mais dont les distances entre amas ne sont pas interprétables. Pour l'ACP, la variance expliquée cumulée donne la part d'information conservée en 2D.

## Q6–Q8 — Détection d'anomalies sur D4 (algorithme 24)

**Construction du benchmark (à signaler dans votre rapport).** Les données sont réelles, mais le **taux d'anomalies est construit** : on prend les tumeurs **bénignes** comme observations « normales » et on y mélange un petit nombre de tumeurs **malignes** (environ 5 %), traitées comme anomalies. C'est un protocole courant, mais le taux d'anomalies n'est pas celui d'un contexte réel. Les étiquettes ne servent **qu'à évaluer** ; aucun modèle ne les voit. On répète le tirage des malignes sur 10 graines pour mesurer la variabilité.

```python
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.covariance import EllipticEnvelope
from sklearn.datasets import load_breast_cancer
bc = load_breast_cancer(); Xb, yb = bc.data, bc.target                    # yb : 0 = malin, 1 = bénin
benins, malins = Xb[yb==1], Xb[yb==0]; n_anom = int(round(.05*len(benins)/.95))
print('bénins :', len(benins), '| malins disponibles :', len(malins), '| anomalies injectées par tirage :', n_anom, f'({100*n_anom/(len(benins)+n_anom):.1f} %)')
def scores_anomalie(nom, X):
    Xs = StandardScaler().fit_transform(X)
    if nom=='Isolation Forest':   return -IsolationForest(n_estimators=200,random_state=GRAINE).fit(Xs).score_samples(Xs)
    if nom=='LOF (voisins=20)':   return -LocalOutlierFactor(n_neighbors=20).fit(Xs).negative_outlier_factor_
    if nom=='Distance de Mahalanobis (référence)': return -EllipticEnvelope(random_state=GRAINE,support_fraction=.9).fit(Xs).score_samples(Xs)
res = {n:{'ROC-AUC':[],'PR-AUC':[],'Précision@k':[]} for n in ('Isolation Forest','LOF (voisins=20)','Distance de Mahalanobis (référence)')}
for graine in range(10):
    r = np.random.default_rng(graine); sel = r.choice(len(malins),n_anom,replace=False)
    X = np.vstack([benins,malins[sel]]); lab = np.r_[np.zeros(len(benins)),np.ones(n_anom)]
    for n in res:
        s = scores_anomalie(n,X); res[n]['ROC-AUC'].append(roc_auc_score(lab,s)); res[n]['PR-AUC'].append(average_precision_score(lab,s))
        res[n]['Précision@k'].append(lab[np.argsort(-s)[:n_anom]].mean())
TAB_ANO = pd.DataFrame({n:{f'{k} (moy ± écart-type)':f'{np.mean(v):.3f} ± {np.std(v):.3f}' for k,v in d.items()} for n,d in res.items()}).T
RESULTATS['anomalies'] = TAB_ANO; display(TAB_ANO)
```

**Interprétation (à adapter à vos chiffres).** La *précision@k* est la part de vraies anomalies parmi les $k$ points les mieux classés comme atypiques ($k$ = nombre d'anomalies injectées) : c'est une mesure de **qualité des alertes**. La PR-AUC est la plus exigeante quand les anomalies sont rares ; comparez-la à son niveau aléatoire (≈ le taux d'anomalies, 5 %). Référence : distance de Mahalanobis ; méthodes avancées : Isolation Forest et LOF. L'écart-type sur 10 tirages indique si une différence entre méthodes est **réelle ou seulement due au tirage**. **Limites** : les anomalies (tumeurs malignes) ne sont pas « rares » dans la réalité ; en production il faut aussi fixer un seuil d'alerte et un coût par fausse alerte.

## Q7–Q8 — Renforcement (algorithmes 25 à 29) : environnements **simulés**, pas des datasets

> **Distinction à rappeler dans le rapport** : les trajectoires utilisées ici sont **produites par l'interaction** de l'agent avec un simulateur. Ce ne sont pas des données réelles. Gymnasium ne propose pas de bandit manchot « standard » : le bandit (25, 26) est donc une **simulation Bernoulli écrite par nous**, signalée comme telle ; Q-learning et SARSA (27, 28) utilisent des environnements **publics Gymnasium**.

```python
# ── Bandits (simulation Bernoulli, 3 bras 0,2 / 0,5 / 0,7), 30 graines, 1000 tours ─
P_BRAS = np.array([.2,.5,.7]); T = 1000
def bandit(strategie, graine, eps=.1):
    g = np.random.default_rng(graine); s=np.zeros(3); n=np.zeros(3); total=0.; best=0
    for t in range(T):
        if strategie=='rotation': a = t%3
        else: a = int(g.integers(3)) if g.random()<eps else int(np.argmax(s/np.maximum(n,1)))
        r = float(g.random()<P_BRAS[a]); s[a]+=r; n[a]+=1; total+=r; best += (a==2)
    return total, T*P_BRAS.max()-total, best/T
lignes=[]
for nom,(strat,eps) in {'25. Exploration uniforme (rotation)':('rotation',0),'Glouton pur (ε=0)':('eps',0.),'26. ε-greedy (ε=0,1)':('eps',.1),'ε-greedy (ε=0,3)':('eps',.3)}.items():
    out = np.array([bandit(strat,k,eps) for k in range(30)])
    lignes.append({'Stratégie':nom,'Récompense cumulée (moy ± ET)':f'{out[:,0].mean():.0f} ± {out[:,0].std():.0f}','Regret moyen':round(out[:,1].mean(),1),'Part du meilleur bras':round(out[:,2].mean(),3)})
TAB_BAN = pd.DataFrame(lignes).set_index('Stratégie'); RESULTATS['bandits']=TAB_BAN; display(TAB_BAN)
```

```python
# ── Q-learning vs SARSA : environnements publics Gymnasium, 5 graines ─────────
import gymnasium as gym
def entrainer(algo, env_id, episodes, graine, alpha, gamma, eps0, eps1, frac, **kw):
    env = gym.make(env_id, **kw); g = np.random.default_rng(graine); nS, nA = env.observation_space.n, env.action_space.n; Q = np.zeros((nS,nA)); retours=[]
    def choisir(s,e):
        return int(g.integers(nA)) if g.random()<e else int(g.choice(np.flatnonzero(Q[s]==Q[s].max())))   # départage aléatoire des égalités
    for ep in range(episodes):
        e = max(eps1, eps0-(eps0-eps1)*ep/(frac*episodes)); s,_ = env.reset(seed=graine) if ep==0 else env.reset(); a = choisir(s,e); G=0.
        for _ in range(500):
            s2,r,term,trunc,_ = env.step(a); G+=r; a2 = choisir(s2,e)
            cible = r + (0. if term else gamma*(Q[s2].max() if algo=='qlearning' else Q[s2,a2]))
            Q[s,a] += alpha*(cible-Q[s,a]); s,a = s2,a2
            if term or trunc: break
        retours.append(G)
    env.close(); return Q, np.array(retours)
def evaluer(Q, env_id, n, graine, succes, **kw):
    env = gym.make(env_id, **kw); G=[]; ok=[]
    for ep in range(n):
        s,_ = env.reset(seed=graine) if ep==0 else env.reset(); tot=0.; term=False
        for _ in range(500):
            s,r,term,trunc,_ = env.step(int(np.argmax(Q[s]))); tot+=r
            if term or trunc: break
        G.append(tot); ok.append(succes(term,tot))
    env.close(); return np.mean(G), np.mean(ok)
def id_cliff():
    for i in ('CliffWalking-v1','CliffWalking-v0'):
        try: gym.make(i).close(); return i
        except Exception: pass
    raise RuntimeError('CliffWalking introuvable dans cette version de Gymnasium')
ID_CLIFF = id_cliff(); print('Environnement falaise utilisé :', ID_CLIFF, '| gymnasium', gym.__version__)
CONFIGS = {
 f'{ID_CLIFF} (déterministe, risque de falaise)': dict(env_id=ID_CLIFF,episodes=500,alpha=.5,gamma=1.,eps0=.1,eps1=.1,frac=1.,succes=lambda term,tot: term,kw={},n_eval=20),
 'FrozenLake-v1 (glissant, stochastique)':          dict(env_id='FrozenLake-v1',episodes=8000,alpha=.1,gamma=.99,eps0=1.,eps1=.05,frac=.6,succes=lambda term,tot: term and tot>0,kw={'is_slippery':True},n_eval=500),
}
lignes=[]
for nom_env,c in CONFIGS.items():
    for algo in ('qlearning','sarsa'):
        t0=time.time(); res=[]
        for graine in range(5):
            Q,ret = entrainer(algo,c['env_id'],c['episodes'],graine,c['alpha'],c['gamma'],c['eps0'],c['eps1'],c['frac'],**c['kw'])
            Gg,ok = evaluer(Q,c['env_id'],c['n_eval'],graine,c['succes'],**c['kw']); res.append((ret[-100:].mean(),Gg,ok))
        res=np.array(res)
        lignes.append({'Environnement':nom_env,'Algorithme':'27. Q-learning' if algo=='qlearning' else '28. SARSA','Récompense d\'entraînement (100 derniers ép., moy ± ET)':f'{res[:,0].mean():.2f} ± {res[:,0].std():.2f}',
                       'Récompense gloutonne (moy ± ET)':f'{res[:,1].mean():.2f} ± {res[:,1].std():.2f}','Taux de réussite gloutons (moy ± ET)':f'{res[:,2].mean():.2f} ± {res[:,2].std():.2f}','Temps (s)':round(time.time()-t0,1)})
TAB_RL = pd.DataFrame(lignes).set_index(['Environnement','Algorithme']); RESULTATS['renforcement']=TAB_RL; display(TAB_RL)
```

**Interprétation (à adapter à vos chiffres).**
- **Bandits** : l'exploration uniforme a un regret qui croît linéairement (un tiers des tirages va au pire bras) ; le glouton pur peut se bloquer sur un mauvais bras (grand écart-type) ; ε-greedy avec un ε modéré obtient le meilleur compromis, un ε trop grand continue d'explorer inutilement. **Simulation, pas données réelles.**
- **CliffWalking** : c'est l'exemple classique où **SARSA (on-policy) et Q-learning (off-policy) diffèrent**. Pendant l'entraînement avec ε constant, le Q-learning longe la falaise (chemin optimal mais risqué) et subit des chutes qui dégradent sa récompense d'entraînement, tandis que SARSA contourne le danger. En évaluation gloutonne, le Q-learning suit le chemin court. Comparez les deux colonnes de récompense pour le vérifier sur **vos** résultats.

```python
# Régression linéaire par descente de gradient : données univariées
x=np.linspace(-1,1,100); y=2+3*x
poids=0.;biais=0.;pas=.1
for iteration in range(500):
    prediction=poids*x+biais
    erreur=prediction-y
    poids-=pas*2*np.mean(erreur*x)
    biais-=pas*2*np.mean(erreur)
print('Paramètres appris manuellement:',poids,biais)
from sklearn.linear_model import LinearRegression
print('Référence sklearn:',LinearRegression().fit(x.reshape(-1,1),y).coef_,LinearRegression().fit(x.reshape(-1,1),y).intercept_)
```

```python
# K-means manuel : assignation et recalcul jusqu'à stabilité
valeurs=X_clu.copy(); centroids=valeurs[np.random.default_rng(GRAINE).choice(len(valeurs),size=3,replace=False)].copy()
for iteration in range(100):
    distances=((valeurs[:,None,:]-centroids[None,:,:])**2).sum(axis=2)
    groupes=distances.argmin(axis=1)
    nouveaux=np.array([valeurs[groupes==k].mean(axis=0) if (groupes==k).any() else centroids[k] for k in range(3)])
    if np.allclose(nouveaux,centroids):break
    centroids=nouveaux
print('Centroïdes manuels:',centroids)
from sklearn.cluster import KMeans
print('Centroïdes sklearn (ordre arbitraire):',KMeans(n_clusters=3,n_init=10,random_state=GRAINE).fit(valeurs).cluster_centers_)
```
